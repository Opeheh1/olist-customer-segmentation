```python

# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import os

# Display all columns when inspecting datasets
pd.set_option("display.max_columns", None)

# Improve numerical display
pd.set_option("display.float_format", "{:,.2f}".format)


# ============================================================
# 2. DEFINE DATASET DIRECTORY
# ============================================================

# Update this path to the folder containing your Olist CSV files.

DATA_PATH = r"C:/Users/HP/Documents/Nexford_Project_File"


# ============================================================
# 3. DEFINE THE SIX DATA FILES
# ============================================================

file_names = {
    "customers": "C:/Users/HP/Documents/Nexford_Project_File/olist_customers_dataset.csv",
    "orders": "C:/Users/HP/Documents/Nexford_Project_File/olist_orders_dataset.csv",
    "order_items": "C:/Users/HP/Documents/Nexford_Project_File/olist_order_items_dataset.csv",
    "payments": "C:/Users/HP/Documents/Nexford_Project_File/olist_order_payments_dataset.csv",
    "reviews": "C:/Users/HP/Documents/Nexford_Project_File/olist_order_reviews_dataset.csv",
    "products": "C:/Users/HP/Documents/Nexford_Project_File/olist_products_dataset.csv"
}


# ============================================================
# 4. LOAD THE DATASETS
# ============================================================

datasets = {}

for name, file_name in file_names.items():

    file_path = os.path.join(DATA_PATH, file_name)

    try:
        datasets[name] = pd.read_csv(file_path)
        print(f"{name.upper()} dataset loaded successfully.")

    except FileNotFoundError:
        print(f"ERROR: {file_name} was not found.")

    except Exception as e:
        print(f"ERROR loading {file_name}: {e}")


# ============================================================
# 5. DISPLAY DATASET DIMENSIONS
# ============================================================

print("\nDATASET DIMENSIONS")
print("=" * 60)

for name, df in datasets.items():

    rows, columns = df.shape

    print(f"{name.upper():15} | Rows: {rows:>10,} | Columns: {columns:>5}")
```

```python
# ============================================================
# 6. DATASET STRUCTURE AND SAMPLE RECORDS
# ============================================================

for name, df in datasets.items():

    print("\n" + "=" * 80)
    print(f"{name.upper()} DATASET")
    print("=" * 80)

    # 1. Display column names and data types
    print("\nCOLUMN NAMES AND DATA TYPES:")
    print(df.dtypes)

    # 2. Display the first three records
    print("\nFIRST THREE RECORDS:")
    display(df.head(3))

    # 3. Display basic structural information
    print("\nDATASET INFORMATION:")
    print(f"Number of rows: {df.shape[0]:,}")
    print(f"Number of columns: {df.shape[1]}")
    print(f"Memory usage: {df.memory_usage(deep=True).sum() / (1024**2):.2f} MB")
```

```python
# ============================================================
# MISSING VALUES AND DUPLICATE ANALYSIS
# ============================================================

# ------------------------------------------------------------
# 7. MISSING VALUE ANALYSIS
# ------------------------------------------------------------

print("=" * 80)
print("MISSING VALUE ANALYSIS")
print("=" * 80)

for name, df in datasets.items():

    print(f"\n{name.upper()} DATASET")
    print("-" * 60)

    missing_count = df.isnull().sum()
    missing_percentage = (missing_count / len(df)) * 100

    missing_summary = pd.DataFrame({
        "Missing Count": missing_count,
        "Missing Percentage (%)": missing_percentage
    })

    # Display only columns containing missing values
    missing_summary = missing_summary[
        missing_summary["Missing Count"] > 0
    ].sort_values("Missing Count", ascending=False)

    if missing_summary.empty:
        print("No missing values found.")
    else:
        display(missing_summary)


# ------------------------------------------------------------
# 8. DUPLICATE ROW ANALYSIS
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("DUPLICATE ROW ANALYSIS")
print("=" * 80)

for name, df in datasets.items():

    duplicate_count = df.duplicated().sum()

    print(
        f"{name.upper():15} | "
        f"Duplicate rows: {duplicate_count:,}"
    )


# ------------------------------------------------------------
# 9. DUPLICATE IDENTIFIER ANALYSIS
# ------------------------------------------------------------

# Candidate identifiers and composite keys based on
# the structure of the Olist datasets.

candidate_keys = {
    "customers": ["customer_id"],
    "orders": ["order_id"],
    "order_items": ["order_id", "order_item_id"],
    "payments": ["order_id", "payment_sequential"],
    "reviews": ["review_id"],
    "products": ["product_id"]
}

print("\n" + "=" * 80)
print("DUPLICATE KEY ANALYSIS")
print("=" * 80)

for name, keys in candidate_keys.items():

    df = datasets[name]

    duplicate_keys = df.duplicated(subset=keys).sum()

    print(
        f"{name.upper():15} | "
        f"Key: {', '.join(keys):35} | "
        f"Duplicate combinations: {duplicate_keys:,}"
    )
```

```python
# ============================================================
# INVESTIGATE DATA QUALITY PATTERNS
# ============================================================

# ------------------------------------------------------------
# 10. INVESTIGATE DUPLICATED REVIEW IDs
# ------------------------------------------------------------

reviews = datasets["reviews"]

print("=" * 80)
print("REVIEW ID INVESTIGATION")
print("=" * 80)

# Number of distinct review IDs
print("Total review records:", len(reviews))
print("Unique review IDs:", reviews["review_id"].nunique())

# Number of records associated with duplicated review IDs
duplicated_review_ids = reviews[
    reviews.duplicated(subset=["review_id"], keep=False)
].sort_values("review_id")

print("\nRecords with duplicated review IDs:")
display(duplicated_review_ids.head(15))

# Number of duplicated review ID groups
print(
    "\nNumber of duplicated review ID groups:",
    reviews.loc[
        reviews.duplicated("review_id", keep=False),
        "review_id"
    ].nunique()
)

# Check whether duplicated review IDs belong to different orders
review_id_order_counts = (
    reviews.groupby("review_id")["order_id"].nunique()
)

print(
    "\nDuplicated review IDs linked to multiple orders:",
    (review_id_order_counts > 1).sum()
)


# ------------------------------------------------------------
# 11. CHECK WHETHER ORDERS HAVE MULTIPLE REVIEW RECORDS
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("REVIEW RECORDS PER ORDER")
print("=" * 80)

reviews_per_order = reviews.groupby("order_id").size()

print("Total distinct orders reviewed:", reviews["order_id"].nunique())
print("Orders with multiple review records:", (reviews_per_order > 1).sum())
print("Maximum reviews recorded for one order:", reviews_per_order.max())

print("\nDistribution of review records per order:")
display(reviews_per_order.value_counts().sort_index().rename("Number of Orders"))


# ------------------------------------------------------------
# 12. INVESTIGATE MISSING ORDER DATES BY ORDER STATUS
# ------------------------------------------------------------

orders = datasets["orders"]

print("\n" + "=" * 80)
print("MISSING ORDER DATES BY ORDER STATUS")
print("=" * 80)

date_columns = [
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date"
]

for column in date_columns:

    print(f"\nMissing values in {column}:")

    missing_by_status = pd.crosstab(
        orders["order_status"],
        orders[column].isna(),
        margins=True
    )

    missing_by_status.columns = ["Available", "Missing", "Total"]

    display(missing_by_status)


# ------------------------------------------------------------
# 13. INVESTIGATE MISSING PRODUCT INFORMATION
# ------------------------------------------------------------

products = datasets["products"]

print("\n" + "=" * 80)
print("PRODUCT MISSING VALUE PATTERNS")
print("=" * 80)

product_missing_columns = [
    "product_category_name",
    "product_name_lenght",
    "product_description_lenght",
    "product_photos_qty",
    "product_weight_g",
    "product_length_cm",
    "product_height_cm",
    "product_width_cm"
]

# Count missing values in each product record
products["missing_field_count"] = (
    products[product_missing_columns].isna().sum(axis=1)
)

print("Products with at least one missing field:",
      (products["missing_field_count"] > 0).sum())

print("\nDistribution of missing fields per product:")
display(
    products["missing_field_count"]
    .value_counts()
    .sort_index()
    .rename("Number of Products")
)

# Check whether the four descriptive fields are missing together
descriptive_columns = [
    "product_category_name",
    "product_name_lenght",
    "product_description_lenght",
    "product_photos_qty"
]

all_descriptive_missing = (
    products[descriptive_columns].isna().all(axis=1)
)

print(
    "\nProducts missing all four descriptive fields:",
    all_descriptive_missing.sum()
)

# Check products with missing physical measurements
physical_columns = [
    "product_weight_g",
    "product_length_cm",
    "product_height_cm",
    "product_width_cm"
]

print(
    "\nProducts missing any physical measurement:",
    products[physical_columns].isna().any(axis=1).sum()
)

# Show examples of affected products
print("\nSample products with missing information:")
display(
    products.loc[
        products["missing_field_count"] > 0,
        ["product_id"] + product_missing_columns
    ].head(10)
)

# Remove temporary exploration column
products.drop(columns=["missing_field_count"], inplace=True)
```

```python
# ============================================================
# DATASET RELATIONSHIP AND KEY INTEGRITY ANALYSIS
# ============================================================

customers = datasets["customers"]
orders = datasets["orders"]
order_items = datasets["order_items"]
payments = datasets["payments"]
reviews = datasets["reviews"]
products = datasets["products"]


# ------------------------------------------------------------
# 1. CHECK KEY UNIQUENESS
# ------------------------------------------------------------

print("=" * 80)
print("KEY UNIQUENESS")
print("=" * 80)

key_checks = {
    "Customers - customer_id": (customers, ["customer_id"]),
    "Customers - customer_unique_id": (customers, ["customer_unique_id"]),
    "Orders - order_id": (orders, ["order_id"]),
    "Order items - composite key": (
        order_items, ["order_id", "order_item_id"]
    ),
    "Payments - composite key": (
        payments, ["order_id", "payment_sequential"]
    ),
    "Products - product_id": (products, ["product_id"])
}

for label, (df, keys) in key_checks.items():

    unique_count = df.drop_duplicates(subset=keys).shape[0]
    duplicate_count = len(df) - unique_count

    print(
        f"{label:40} | "
        f"Unique combinations: {unique_count:,} | "
        f"Duplicates: {duplicate_count:,}"
    )


# ------------------------------------------------------------
# 2. CHECK FOREIGN KEY MATCHING
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("FOREIGN KEY INTEGRITY")
print("=" * 80)

relationship_checks = [
    (
        "Orders -> Customers",
        orders["customer_id"],
        customers["customer_id"]
    ),
    (
        "Order items -> Orders",
        order_items["order_id"],
        orders["order_id"]
    ),
    (
        "Order items -> Products",
        order_items["product_id"],
        products["product_id"]
    ),
    (
        "Payments -> Orders",
        payments["order_id"],
        orders["order_id"]
    ),
    (
        "Reviews -> Orders",
        reviews["order_id"],
        orders["order_id"]
    )
]

for label, child_values, parent_values in relationship_checks:

    unmatched = (~child_values.isin(parent_values)).sum()

    print(
        f"{label:30} | "
        f"Unmatched records: {unmatched:,}"
    )


# ------------------------------------------------------------
# 3. ORDERS PER CUSTOMER
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("CUSTOMER ORDER FREQUENCY")
print("=" * 80)

# Attach the stable customer identifier to each order
customer_orders = orders.merge(
    customers[["customer_id", "customer_unique_id"]],
    on="customer_id",
    how="left",
    validate="one_to_one"
)

orders_per_customer = (
    customer_orders.groupby("customer_unique_id")["order_id"]
    .nunique()
)

print("Unique customers:", orders_per_customer.shape[0])
print("Customers with one order:", (orders_per_customer == 1).sum())
print("Customers with multiple orders:", (orders_per_customer > 1).sum())
print("Maximum orders by one customer:", orders_per_customer.max())

print("\nDistribution of orders per customer:")
display(
    orders_per_customer.value_counts()
    .sort_index()
    .rename("Number of Customers")
    .to_frame()
)


# ------------------------------------------------------------
# 4. ITEMS PER ORDER
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("ORDER ITEM FREQUENCY")
print("=" * 80)

items_per_order = order_items.groupby("order_id").size()

print("Orders with items:", items_per_order.shape[0])
print("Orders with multiple items:", (items_per_order > 1).sum())
print("Maximum items in one order:", items_per_order.max())

print("\nDistribution of items per order:")
display(
    items_per_order.value_counts()
    .sort_index()
    .rename("Number of Orders")
    .to_frame()
)


# ------------------------------------------------------------
# 5. PAYMENTS PER ORDER
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("PAYMENT FREQUENCY")
print("=" * 80)

payments_per_order = payments.groupby("order_id").size()

print("Orders with payments:", payments_per_order.shape[0])
print("Orders with multiple payment records:",
      (payments_per_order > 1).sum())
print("Maximum payment records per order:", payments_per_order.max())

print("\nDistribution of payments per order:")
display(
    payments_per_order.value_counts()
    .sort_index()
    .rename("Number of Orders")
    .to_frame()
)


# ------------------------------------------------------------
# 6. REVIEWS PER ORDER
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("REVIEW FREQUENCY")
print("=" * 80)

reviews_per_order = reviews.groupby("order_id").size()

print("Orders with reviews:", reviews_per_order.shape[0])
print("Orders with multiple reviews:",
      (reviews_per_order > 1).sum())
print("Maximum reviews per order:", reviews_per_order.max())


# ------------------------------------------------------------
# 7. SUMMARY OF DATASET RELATIONSHIPS
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("RELATIONSHIP SUMMARY")
print("=" * 80)

print("Customers:", customers["customer_id"].nunique())
print("Orders:", orders["order_id"].nunique())
print("Products:", products["product_id"].nunique())
print("Order item records:", len(order_items))
print("Payment records:", len(payments))
print("Review records:", len(reviews))
print("Unique customer identities:",
      customers["customer_unique_id"].nunique())
```

```python
# EXPLORE ORDER STATUS AND DATE COVERAGE
# ============================================================

import pandas as pd
import numpy as np

# Work with a temporary copy so the original orders dataset
# remains unchanged during exploration.
orders_analysis = orders.copy()

# Convert date columns temporarily for analysis.
date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in date_columns:
    orders_analysis[col] = pd.to_datetime(
        orders_analysis[col],
        errors="coerce"
    )

print("=" * 65)
print("ORDER STATUS DISTRIBUTION")
print("=" * 65)

status_summary = (
    orders_analysis["order_status"]
    .value_counts(dropna=False)
    .rename_axis("Order Status")
    .reset_index(name="Number of Orders")
)

status_summary["Percentage"] = (
    status_summary["Number of Orders"] /
    len(orders_analysis) * 100
).round(2)

print(status_summary.to_string(index=False))


print("\n" + "=" * 65)
print("PURCHASE DATE COVERAGE")
print("=" * 65)

purchase_dates = orders_analysis["order_purchase_timestamp"]

print("Earliest purchase date:", purchase_dates.min())
print("Latest purchase date:", purchase_dates.max())
print("Missing purchase dates:", purchase_dates.isna().sum())

# Yearly order distribution
yearly_orders = (
    orders_analysis
    .dropna(subset=["order_purchase_timestamp"])
    .assign(
        purchase_year=lambda x: x["order_purchase_timestamp"].dt.year
    )
    .groupby("purchase_year")["order_id"]
    .nunique()
)

print("\nOrders by year:")
print(yearly_orders.to_string())


# Monthly order distribution
monthly_orders = (
    orders_analysis
    .dropna(subset=["order_purchase_timestamp"])
    .assign(
        purchase_month=lambda x:
        x["order_purchase_timestamp"].dt.to_period("M").astype(str)
    )
    .groupby("purchase_month")["order_id"]
    .nunique()
)

print("\nFirst 10 months:")
print(monthly_orders.head(10).to_string())

print("\nLast 10 months:")
print(monthly_orders.tail(10).to_string())


# ============================================================
# EXPLORE ORDER ITEM SPENDING AND FREIGHT
# ============================================================

print("\n" + "=" * 65)
print("ORDER ITEM PRICE AND FREIGHT DISTRIBUTION")
print("=" * 65)

for col in ["price", "freight_value"]:
    print(f"\n{col.upper()} SUMMARY:")
    print(order_items[col].describe().round(2))

    print("Zero values:", (order_items[col] == 0).sum())
    print("Negative values:", (order_items[col] < 0).sum())
    print("Missing values:", order_items[col].isna().sum())


# Aggregate item spending and freight by order.
# This is an exploratory aggregation, not a modification
# of the original order_items dataset.

order_item_summary = (
    order_items
    .groupby("order_id")
    .agg(
        item_count=("order_item_id", "count"),
        item_spending=("price", "sum"),
        freight_spending=("freight_value", "sum")
    )
)

print("\n" + "=" * 65)
print("AGGREGATED ORDER SPENDING")
print("=" * 65)

print(order_item_summary.describe().round(2))


# ============================================================
# EXPLORE PAYMENT BEHAVIOUR
# ============================================================

print("\n" + "=" * 65)
print("PAYMENT VALUE DISTRIBUTION")
print("=" * 65)

print(payments["payment_value"].describe().round(2))

print("\nZero payment values:",
      (payments["payment_value"] == 0).sum())

print("Negative payment values:",
      (payments["payment_value"] < 0).sum())

print("\n" + "=" * 65)
print("PAYMENT TYPE DISTRIBUTION")
print("=" * 65)

payment_type_summary = (
    payments["payment_type"]
    .value_counts(dropna=False)
    .rename_axis("Payment Type")
    .reset_index(name="Number of Records")
)

payment_type_summary["Percentage"] = (
    payment_type_summary["Number of Records"] /
    len(payments) * 100
).round(2)

print(payment_type_summary.to_string(index=False))


# ============================================================
# EXPLORE CUSTOMER REVIEW SCORES
# ============================================================

print("\n" + "=" * 65)
print("REVIEW SCORE DISTRIBUTION")
print("=" * 65)

review_score_summary = (
    reviews["review_score"]
    .value_counts(dropna=False)
    .sort_index()
    .rename_axis("Review Score")
    .reset_index(name="Number of Reviews")
)

review_score_summary["Percentage"] = (
    review_score_summary["Number of Reviews"] /
    len(reviews) * 100
).round(2)

print(review_score_summary.to_string(index=False))

print("\nAverage review score:",
      round(reviews["review_score"].mean(), 2))

print("Missing review scores:",
      reviews["review_score"].isna().sum())

print("Scores outside the expected 1–5 range:",
      (~reviews["review_score"].between(1, 5)).sum())


# ============================================================
# EXPLORE PRODUCT CATEGORIES AND DIMENSIONS
# ============================================================

print("\n" + "=" * 65)
print("PRODUCT CATEGORY DISTRIBUTION")
print("=" * 65)

print("Number of distinct product categories:",
      products["product_category_name"].nunique())

print("Products with missing category:",
      products["product_category_name"].isna().sum())

print("\nTop 10 product categories:")

print(
    products["product_category_name"]
    .value_counts()
    .head(10)
    .to_string()
)


# Check for zero or negative product measurements.
measurement_columns = [
    "product_weight_g",
    "product_length_cm",
    "product_height_cm",
    "product_width_cm"
]

print("\n" + "=" * 65)
print("PRODUCT MEASUREMENT QUALITY")
print("=" * 65)

for col in measurement_columns:
    print(f"\n{col}")
    print("Missing:", products[col].isna().sum())
    print("Zero values:", (products[col] == 0).sum())
    print("Negative values:", (products[col] < 0).sum())


print("\n" + "=" * 65)
print("STEP 6 EXPLORATION COMPLETE")
print("=" * 65)
```

```python
# ============================================================
# FINAL DATA EXPLORATION AND QUALITY CHECKS
# ============================================================

# ------------------------------------------------------------
# 1. CHECK ORDERS WITHOUT ITEM OR PAYMENT RECORDS
# ------------------------------------------------------------

print("=" * 65)
print("ORDERS WITHOUT ITEM OR PAYMENT RECORDS")
print("=" * 65)

# Identify order IDs that have item and payment records.
orders_with_items = set(order_items["order_id"].unique())
orders_with_payments = set(payments["order_id"].unique())

# Create a temporary order-level analysis table.
order_completeness = orders[
    ["order_id", "order_status"]
].copy()

order_completeness["has_items"] = (
    order_completeness["order_id"].isin(orders_with_items)
)

order_completeness["has_payment"] = (
    order_completeness["order_id"].isin(orders_with_payments)
)

print("Orders without item records:",
      (~order_completeness["has_items"]).sum())

print("Orders without payment records:",
      (~order_completeness["has_payment"]).sum())

print("\nItem record availability by order status:")

print(pd.crosstab(
    order_completeness["order_status"],
    order_completeness["has_items"],
    margins=True
))

print("\nPayment record availability by order status:")

print(pd.crosstab(
    order_completeness["order_status"],
    order_completeness["has_payment"],
    margins=True
))


# ------------------------------------------------------------
# 2. CHECK EXTREME VALUES USING THE IQR METHOD
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("EXTREME VALUE CHECKS")
print("=" * 65)

# These are exploratory flags, not automatic deletion rules.
# An extreme value may represent a genuine purchase.

outlier_columns = {
    "Item price": order_items["price"],
    "Freight value": order_items["freight_value"],
    "Payment value": payments["payment_value"],
    "Order item spending": order_item_summary["item_spending"],
    "Order freight spending": order_item_summary["freight_spending"]
}

for name, series in outlier_columns.items():

    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)
    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outliers = (
        (series < lower_bound) |
        (series > upper_bound)
    )

    print(f"\n{name}")
    print("Lower IQR boundary:", round(lower_bound, 2))
    print("Upper IQR boundary:", round(upper_bound, 2))
    print("Potential outliers:", int(outliers.sum()))
    print("Percentage flagged:",
          round(outliers.mean() * 100, 2), "%")


# ------------------------------------------------------------
# 3. CHECK ORDER DATE SEQUENCE
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("ORDER DATE SEQUENCE CHECK")
print("=" * 65)

date_check = orders_analysis.copy()

# Check whether approval happened before purchase.
date_check["approval_before_purchase"] = (
    date_check["order_approved_at"].notna() &
    (date_check["order_approved_at"] <
     date_check["order_purchase_timestamp"])
)

# Check whether carrier delivery happened before approval.
date_check["carrier_before_approval"] = (
    date_check["order_delivered_carrier_date"].notna() &
    date_check["order_approved_at"].notna() &
    (date_check["order_delivered_carrier_date"] <
     date_check["order_approved_at"])
)

# Check whether customer delivery happened before carrier handover.
date_check["delivery_before_carrier"] = (
    date_check["order_delivered_customer_date"].notna() &
    date_check["order_delivered_carrier_date"].notna() &
    (date_check["order_delivered_customer_date"] <
     date_check["order_delivered_carrier_date"])
)

# Check whether actual customer delivery happened after
# the estimated delivery date.
date_check["delivered_after_estimate"] = (
    date_check["order_delivered_customer_date"].notna() &
    date_check["order_estimated_delivery_date"].notna() &
    (date_check["order_delivered_customer_date"] >
     date_check["order_estimated_delivery_date"])
)

checks = [
    "approval_before_purchase",
    "carrier_before_approval",
    "delivery_before_carrier",
    "delivered_after_estimate"
]

for col in checks:
    print(f"{col}: {date_check[col].sum()}")


# ------------------------------------------------------------
# 4. COMPARE ORDER ITEM TOTALS WITH PAYMENT TOTALS
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("ORDER ITEM AND PAYMENT VALUE COMPARISON")
print("=" * 65)

# Aggregate payment records to one row per order.
payment_by_order = (
    payments.groupby("order_id")
    .agg(
        total_payment=("payment_value", "sum"),
        payment_record_count=("payment_sequential", "count")
    )
)

# Combine the already aggregated item values with payments.
# This is a one-row-per-order comparison.
financial_comparison = order_item_summary.join(
    payment_by_order,
    how="inner"
)

financial_comparison["item_plus_freight"] = (
    financial_comparison["item_spending"] +
    financial_comparison["freight_spending"]
)

financial_comparison["payment_difference"] = (
    financial_comparison["total_payment"] -
    financial_comparison["item_plus_freight"]
)

print("Orders included in comparison:",
      len(financial_comparison))

print("\nFinancial comparison summary:")

print(
    financial_comparison[
        [
            "item_spending",
            "freight_spending",
            "item_plus_freight",
            "total_payment",
            "payment_difference"
        ]
    ].describe().round(2)
)

print("\nOrders where payment differs from item plus freight:",
      (financial_comparison["payment_difference"].abs() > 0.01).sum())


# ------------------------------------------------------------
# 5. FINAL EXPLORATION SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("FINAL EXPLORATION CHECKS COMPLETED")
print("=" * 65)
```
