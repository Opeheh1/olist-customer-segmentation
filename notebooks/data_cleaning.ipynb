```python

# ============================================================
# OLIST CUSTOMER SEGMENTATION PROJECT

# STAGE 1: DATA SOURCES
# STAGE 2: DATA INGESTION
# ============================================================

import pandas as pd
from pathlib import Path
import logging

# ------------------------------------------------------------
# 1. CONFIGURE THE PROJECT
# ------------------------------------------------------------

# Set the folder containing the original Olist CSV files.
# Replace this placeholder with your actual dataset folder.

DATA_PATH = Path(r"C:/Users/HP/Documents/Nexford_Project_File")

# Define the six required data sources.
DATA_SOURCES = {
    "customers": "C:/Users/HP/Documents/Nexford_Project_File/olist_customers_dataset.csv",
    "orders": "C:/Users/HP/Documents/Nexford_Project_File/olist_orders_dataset.csv",
    "order_items": "C:/Users/HP/Documents/Nexford_Project_File/olist_order_items_dataset.csv",
    "payments": "C:/Users/HP/Documents/Nexford_Project_File/olist_order_payments_dataset.csv",
    "reviews": "C:/Users/HP/Documents/Nexford_Project_File/olist_order_reviews_dataset.csv",
    "products": "C:/Users/HP/Documents/Nexford_Project_File/olist_products_dataset.csv"
}

# ------------------------------------------------------------
# 2. CONFIGURE LOGGING
# ------------------------------------------------------------

# Logging provides a record of ingestion activities.
# This is useful for pipeline monitoring and troubleshooting.

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s | %(levelname)s | %(message)s"
)

logger = logging.getLogger("olist_data_pipeline")

# ------------------------------------------------------------
# 3. CHECK DATA SOURCE AVAILABILITY
# ------------------------------------------------------------

print("=" * 65)
print("DATA SOURCE AVAILABILITY CHECK")
print("=" * 65)

missing_files = []

for dataset_name, filename in DATA_SOURCES.items():

    file_path = DATA_PATH / filename

    if file_path.exists():
        print(f"[FOUND] {dataset_name}: {filename}")
        logger.info(f"Data source found: {filename}")
    else:
        print(f"[MISSING] {dataset_name}: {filename}")
        missing_files.append(filename)
        logger.error(f"Data source missing: {filename}")

# Stop ingestion if a required source is missing.
if missing_files:
    raise FileNotFoundError(
        f"The following required files were not found: {missing_files}"
    )

# ------------------------------------------------------------
# 4. INGEST THE RAW DATASETS
# ------------------------------------------------------------

# Each CSV is loaded into a dictionary.
# The original files are not modified by this operation.

raw_datasets = {}

for dataset_name, filename in DATA_SOURCES.items():

    file_path = DATA_PATH / filename

    try:
        df = pd.read_csv(
            file_path,
            encoding="utf-8",
            low_memory=False
        )

        raw_datasets[dataset_name] = df

        logger.info(
            f"Successfully ingested {filename} "
            f"| Rows: {df.shape[0]} "
            f"| Columns: {df.shape[1]}"
        )

    except Exception as error:
        logger.exception(
            f"Failed to ingest {filename}: {error}"
        )
        raise

# ------------------------------------------------------------
# 5. DISPLAY INGESTION SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("RAW DATA INGESTION SUMMARY")
print("=" * 65)

ingestion_summary = []

for dataset_name, df in raw_datasets.items():

    ingestion_summary.append({
        "Dataset": dataset_name,
        "Rows": df.shape[0],
        "Columns": df.shape[1],
        "Memory (MB)": round(
            df.memory_usage(deep=True).sum() / (1024 ** 2), 2
        )
    })

ingestion_summary = pd.DataFrame(ingestion_summary)

print(ingestion_summary.to_string(index=False))

# ------------------------------------------------------------
# 6. CREATE WORKING REFERENCES
# ------------------------------------------------------------

# These are separate copies for subsequent cleaning.
# raw_datasets remains unchanged.

customers_raw = raw_datasets["customers"].copy()
orders_raw = raw_datasets["orders"].copy()
order_items_raw = raw_datasets["order_items"].copy()
payments_raw = raw_datasets["payments"].copy()
reviews_raw = raw_datasets["reviews"].copy()
products_raw = raw_datasets["products"].copy()

print("\nIngestion completed successfully.")
print("Raw datasets preserved in 'raw_datasets'.")
print("Working copies created for the cleaning stage.")
```

```python

# ============================================================
# STAGE 3: DATA CLEANING
# STEP 1: STANDARDISE DATA TYPES
# ============================================================

# Create fresh working copies from the original ingested data.
# This ensures the raw datasets remain unchanged.

cleaned_datasets = {
    name: df.copy()
    for name, df in raw_datasets.items()
}

# Create convenient references to the working datasets.
customers_clean = cleaned_datasets["customers"]
orders_clean = cleaned_datasets["orders"]
order_items_clean = cleaned_datasets["order_items"]
payments_clean = cleaned_datasets["payments"]
reviews_clean = cleaned_datasets["reviews"]
products_clean = cleaned_datasets["products"]


# ------------------------------------------------------------
# 1. STANDARDISE IDENTIFIER COLUMNS
# ------------------------------------------------------------

# Identifiers are labels, not numerical measurements.
# Convert them to pandas string type to support consistent
# joins and prevent accidental numerical calculations.

identifier_columns = {
    "customers": [
        "customer_id",
        "customer_unique_id"
    ],
    "orders": [
        "order_id",
        "customer_id"
    ],
    "order_items": [
        "order_id",
        "product_id",
        "seller_id"
    ],
    "payments": [
        "order_id"
    ],
    "reviews": [
        "review_id",
        "order_id"
    ],
    "products": [
        "product_id"
    ]
}

for dataset_name, columns in identifier_columns.items():

    df = cleaned_datasets[dataset_name]

    for col in columns:
        df[col] = df[col].astype("string")


# ------------------------------------------------------------
# 2. STANDARDISE CUSTOMER POSTAL CODE
# ------------------------------------------------------------

# Postal codes are identifiers, not numerical values.
# Convert to string and retain a five-character format.

customers_clean["customer_zip_code_prefix"] = (
    customers_clean["customer_zip_code_prefix"]
    .astype("string")
    .str.zfill(5)
)


# ------------------------------------------------------------
# 3. CONVERT ORDER TIMESTAMP COLUMNS
# ------------------------------------------------------------

order_date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in order_date_columns:

    orders_clean[col] = pd.to_datetime(
        orders_clean[col],
        errors="coerce"
    )


# Convert the shipping limit date in order items.
order_items_clean["shipping_limit_date"] = pd.to_datetime(
    order_items_clean["shipping_limit_date"],
    errors="coerce"
)


# Convert review date columns.
review_date_columns = [
    "review_creation_date",
    "review_answer_timestamp"
]

for col in review_date_columns:

    reviews_clean[col] = pd.to_datetime(
        reviews_clean[col],
        errors="coerce"
    )


# ------------------------------------------------------------
# 4. CHECK STANDARDISED DATA TYPES
# ------------------------------------------------------------

print("=" * 65)
print("STANDARDISED DATA TYPES")
print("=" * 65)

for dataset_name, df in cleaned_datasets.items():

    print(f"\n{dataset_name.upper()}")
    print(df.dtypes.to_string())


# ------------------------------------------------------------
# 5. CHECK DATE CONVERSION RESULTS
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("DATE CONVERSION CHECK")
print("=" * 65)

date_conversion_summary = []

date_columns_by_dataset = {
    "orders": order_date_columns,
    "order_items": ["shipping_limit_date"],
    "reviews": review_date_columns
}

for dataset_name, columns in date_columns_by_dataset.items():

    original_df = raw_datasets[dataset_name]
    cleaned_df = cleaned_datasets[dataset_name]

    for col in columns:

        original_non_missing = original_df[col].notna().sum()
        converted_missing = cleaned_df[col].isna().sum()

        # Values that existed originally but could not be
        # converted into valid dates.
        conversion_failures = (
            original_df[col].notna() &
            cleaned_df[col].isna()
        ).sum()

        date_conversion_summary.append({
            "Dataset": dataset_name,
            "Column": col,
            "Original Non-Missing": original_non_missing,
            "Missing After Conversion": converted_missing,
            "Conversion Failures": conversion_failures
        })

date_conversion_summary = pd.DataFrame(
    date_conversion_summary
)

print(date_conversion_summary.to_string(index=False))


# ------------------------------------------------------------
# 6. GENERATE UPDATED MISSING VALUE REPORT
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("MISSING VALUE REPORT AFTER TYPE STANDARDISATION")
print("=" * 65)

for dataset_name, df in cleaned_datasets.items():

    missing = df.isna().sum()
    missing = missing[missing > 0]

    print(f"\n{dataset_name.upper()}")

    if missing.empty:
        print("No missing values.")
    else:
        missing_report = pd.DataFrame({
            "Missing Count": missing,
            "Missing Percentage": (
                missing / len(df) * 100
            ).round(2)
        })

        print(missing_report.to_string())


print("\nData type standardisation completed.")
print("Original raw datasets remain unchanged.")
```

```python

# ============================================================
# STAGE 3: DATA CLEANING
# STEP 2: MISSING VALUE ASSESSMENT AND HANDLING
# ============================================================

# Our approach:
# 1. Preserve meaningful missing values.
# 2. Investigate missingness in relation to business context.
# 3. Create indicators to identify incomplete records.
# 4. Avoid arbitrary imputation or deletion.


# ------------------------------------------------------------
# 1. INVESTIGATE MISSING ORDER DATES BY ORDER STATUS
# ------------------------------------------------------------

print("=" * 65)
print("MISSING ORDER DATES BY ORDER STATUS")
print("=" * 65)

order_date_missing = (
    orders_clean
    .groupby("order_status")
    .agg(
        total_orders=("order_id", "count"),
        missing_approval=("order_approved_at",
                          lambda x: x.isna().sum()),
        missing_carrier_date=("order_delivered_carrier_date",
                              lambda x: x.isna().sum()),
        missing_delivery_date=("order_delivered_customer_date",
                               lambda x: x.isna().sum())
    )
)

print(order_date_missing.to_string())


# Create flags rather than replacing missing dates.
orders_clean["approval_date_missing"] = (
    orders_clean["order_approved_at"].isna()
)

orders_clean["carrier_date_missing"] = (
    orders_clean["order_delivered_carrier_date"].isna()
)

orders_clean["delivery_date_missing"] = (
    orders_clean["order_delivered_customer_date"].isna()
)


# ------------------------------------------------------------
# 2. INVESTIGATE MISSING REVIEW COMMENTS
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("REVIEW COMMENT COMPLETENESS")
print("=" * 65)

reviews_clean["has_review_title"] = (
    reviews_clean["review_comment_title"].notna()
)

reviews_clean["has_review_message"] = (
    reviews_clean["review_comment_message"].notna()
)

reviews_clean["has_any_review_comment"] = (
    reviews_clean["has_review_title"] |
    reviews_clean["has_review_message"]
)

print("Total review records:", len(reviews_clean))

print("Reviews with a title:",
      reviews_clean["has_review_title"].sum())

print("Reviews with a message:",
      reviews_clean["has_review_message"].sum())

print("Reviews with at least one comment:",
      reviews_clean["has_any_review_comment"].sum())

print("Reviews without any comment:",
      (~reviews_clean["has_any_review_comment"]).sum())

print("Missing review scores:",
      reviews_clean["review_score"].isna().sum())


# ------------------------------------------------------------
# 3. INVESTIGATE MISSING PRODUCT ATTRIBUTES
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("PRODUCT ATTRIBUTE COMPLETENESS")
print("=" * 65)

product_descriptive_columns = [
    "product_category_name",
    "product_name_lenght",
    "product_description_lenght",
    "product_photos_qty"
]

product_measurement_columns = [
    "product_weight_g",
    "product_length_cm",
    "product_height_cm",
    "product_width_cm"
]

# Count missing descriptive attributes per product.
products_clean["missing_descriptive_count"] = (
    products_clean[product_descriptive_columns]
    .isna()
    .sum(axis=1)
)

# Count missing physical measurements per product.
products_clean["missing_measurement_count"] = (
    products_clean[product_measurement_columns]
    .isna()
    .sum(axis=1)
)

print("Products missing descriptive attributes:",
      (products_clean["missing_descriptive_count"] > 0).sum())

print("Products missing physical measurements:",
      (products_clean["missing_measurement_count"] > 0).sum())

print("\nDistribution of missing descriptive attributes:")
print(
    products_clean["missing_descriptive_count"]
    .value_counts()
    .sort_index()
    .to_string()
)

print("\nDistribution of missing physical measurements:")
print(
    products_clean["missing_measurement_count"]
    .value_counts()
    .sort_index()
    .to_string()
)


# ------------------------------------------------------------
# 4. CHECK WHETHER INCOMPLETE PRODUCTS HAVE PURCHASE RECORDS
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("PURCHASE ACTIVITY OF INCOMPLETE PRODUCTS")
print("=" * 65)

# Count item records associated with each product.
product_purchase_counts = (
    order_items_clean
    .groupby("product_id")
    .agg(
        item_records=("order_id", "count"),
        distinct_orders=("order_id", "nunique")
    )
)

# Join purchase counts to product information.
# The relationship is many-to-one from order items to products.

product_usage = products_clean[
    [
        "product_id",
        "product_category_name",
        "missing_descriptive_count",
        "missing_measurement_count"
    ]
].merge(
    product_purchase_counts,
    left_on="product_id",
    right_index=True,
    how="left",
    validate="one_to_one"
)

# Products without purchase records receive zero in these
# derived activity measures, not in their original attributes.
product_usage[["item_records", "distinct_orders"]] = (
    product_usage[["item_records", "distinct_orders"]]
    .fillna(0)
)

incomplete_products = product_usage[
    (product_usage["missing_descriptive_count"] > 0) |
    (product_usage["missing_measurement_count"] > 0)
]

print("Incomplete products:", len(incomplete_products))

print("Incomplete products with purchase records:",
      (incomplete_products["item_records"] > 0).sum())

print("Total item records involving incomplete products:",
      int(incomplete_products["item_records"].sum()))

print("\nIncomplete product activity summary:")
print(
    incomplete_products[
        ["item_records", "distinct_orders"]
    ].describe().round(2)
)


# ------------------------------------------------------------
# 5. GENERATE MISSING VALUE AUDIT
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("MISSING VALUE AUDIT")
print("=" * 65)

missing_audit = []

for dataset_name, df in cleaned_datasets.items():

    for column in df.columns:

        missing_count = df[column].isna().sum()

        if missing_count > 0:

            missing_audit.append({
                "Dataset": dataset_name,
                "Column": column,
                "Missing Count": missing_count,
                "Missing Percentage": round(
                    missing_count / len(df) * 100, 2
                ),
                "Treatment": "Preserved for investigation or optional field"
            })

missing_audit = pd.DataFrame(missing_audit)

print(missing_audit.to_string(index=False))

print("\nMissing-value assessment completed.")
print("Original missing values have not been arbitrarily imputed.")
```

```python

# ============================================================
# STAGE 3: DATA CLEANING
# STEP 3: DUPLICATES AND INVALID VALUE ASSESSMENT
# ============================================================

# ------------------------------------------------------------
# 1. CHECK EXACT DUPLICATE ROWS
# ------------------------------------------------------------

print("=" * 65)
print("EXACT DUPLICATE RECORD CHECK")
print("=" * 65)

for dataset_name, df in cleaned_datasets.items():

    duplicate_count = df.duplicated().sum()

    print(
        f"{dataset_name:15} | "
        f"Exact duplicate rows: {duplicate_count}"
    )


# ------------------------------------------------------------
# 2. CHECK BUSINESS KEY UNIQUENESS
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("BUSINESS KEY DUPLICATE CHECK")
print("=" * 65)

business_keys = {
    "customers": ["customer_id"],
    "orders": ["order_id"],
    "order_items": ["order_id", "order_item_id"],
    "payments": ["order_id", "payment_sequential"],
    "products": ["product_id"]
}

for dataset_name, key_columns in business_keys.items():

    df = cleaned_datasets[dataset_name]

    duplicate_count = df.duplicated(
        subset=key_columns
    ).sum()

    print(
        f"{dataset_name:15} | "
        f"Duplicate key combinations: {duplicate_count}"
    )

# Review ID is deliberately excluded because our exploration
# established that it is reused across different orders.


# ------------------------------------------------------------
# 3. INVESTIGATE REUSED REVIEW IDENTIFIERS
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("REVIEW IDENTIFIER INVESTIGATION")
print("=" * 65)

review_id_summary = (
    reviews_clean
    .groupby("review_id")
    .agg(
        distinct_orders=("order_id", "nunique"),
        review_records=("order_id", "count")
    )
)

reused_review_ids = review_id_summary[
    review_id_summary["distinct_orders"] > 1
]

print("Review IDs linked to multiple orders:",
      len(reused_review_ids))

print("Review records associated with reused IDs:",
      int(reused_review_ids["review_records"].sum()))

print("\nDecision: retain review records and use order_id for aggregation.")


# ------------------------------------------------------------
# 4. CHECK NUMERICAL VALUES FOR INVALID RECORDS
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("NUMERICAL VALUE VALIDATION")
print("=" * 65)

numeric_checks = {
    "order_items": [
        "price",
        "freight_value"
    ],
    "payments": [
        "payment_value",
        "payment_installments"
    ],
    "reviews": [
        "review_score"
    ],
    "products": [
        "product_weight_g",
        "product_length_cm",
        "product_height_cm",
        "product_width_cm",
        "product_photos_qty"
    ]
}

for dataset_name, columns in numeric_checks.items():

    df = cleaned_datasets[dataset_name]

    print(f"\n{dataset_name.upper()}")

    for col in columns:

        print(
            f"{col:30} | "
            f"Negative: {(df[col] < 0).sum():6} | "
            f"Zero: {(df[col] == 0).sum():6}"
        )


# ------------------------------------------------------------
# 5. INVESTIGATE UNUSUAL PAYMENT RECORDS
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("PAYMENT QUALITY INVESTIGATION")
print("=" * 65)

print("Payment types:")
print(payments_clean["payment_type"].value_counts(dropna=False))

print("\nPayment records with zero value:")
print(
    payments_clean[
        payments_clean["payment_value"] == 0
    ].to_string(index=False)
)

print("\nPayment records with not_defined type:")
print(
    payments_clean[
        payments_clean["payment_type"] == "not_defined"
    ].to_string(index=False)
)

print("\nPayment installment distribution:")
print(
    payments_clean["payment_installments"]
    .value_counts()
    .sort_index()
    .to_string()
)


# ------------------------------------------------------------
# 6. INVESTIGATE ZERO PRODUCT MEASUREMENTS
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("ZERO PRODUCT MEASUREMENT INVESTIGATION")
print("=" * 65)

zero_weight_products = products_clean[
    products_clean["product_weight_g"] == 0
]

print("Products with zero weight:",
      len(zero_weight_products))

print("\nDetails of products with zero weight:")

print(
    zero_weight_products[
        [
            "product_id",
            "product_category_name",
            "product_weight_g",
            "product_length_cm",
            "product_height_cm",
            "product_width_cm"
        ]
    ].to_string(index=False)
)


# ------------------------------------------------------------
# 7. CHECK REVIEW SCORE VALIDITY
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("REVIEW SCORE VALIDATION")
print("=" * 65)

invalid_review_scores = reviews_clean[
    ~reviews_clean["review_score"].between(1, 5)
]

print("Invalid review scores:",
      len(invalid_review_scores))


# ------------------------------------------------------------
# 8. SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("DUPLICATE AND INVALID VALUE ASSESSMENT COMPLETED")
print("=" * 65)

print("No records have been deleted or modified by this assessment.")
```

```python

# ============================================================
# STAGE 3: DATA CLEANING
# STEP 4: INVESTIGATE TRANSACTION AND DATE INCONSISTENCIES
# ============================================================

# ------------------------------------------------------------
# 1. INVESTIGATE ZERO-VALUE PAYMENT RECORDS AT ORDER LEVEL
# ------------------------------------------------------------

print("=" * 65)
print("ZERO-VALUE PAYMENT INVESTIGATION")
print("=" * 65)

# Summarize all payment records belonging to each order.
payment_order_summary = (
    payments_clean
    .groupby("order_id")
    .agg(
        total_payment_value=("payment_value", "sum"),
        payment_record_count=("payment_value", "count"),
        zero_payment_records=(
            "payment_value",
            lambda x: (x == 0).sum()
        ),
        positive_payment_records=(
            "payment_value",
            lambda x: (x > 0).sum()
        )
    )
    .reset_index()
)

# Identify orders containing at least one zero-value payment.
zero_payment_orders = payment_order_summary[
    payment_order_summary["zero_payment_records"] > 0
]

print("Orders containing zero-value payment records:",
      len(zero_payment_orders))

print("\nBreakdown of zero-payment orders:")

print(
    zero_payment_orders[
        [
            "payment_record_count",
            "zero_payment_records",
            "positive_payment_records",
            "total_payment_value"
        ]
    ].describe().round(2)
)

# Show whether zero payment records occur alongside positive payments.
print("\nOrders with both zero and positive payment records:",
      len(
          zero_payment_orders[
              zero_payment_orders["positive_payment_records"] > 0
          ]
      ))

print("Orders with zero payment records only:",
      len(
          zero_payment_orders[
              zero_payment_orders["positive_payment_records"] == 0
          ]
      ))


# ------------------------------------------------------------
# 2. CHECK ZERO INSTALLMENT RECORDS
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("ZERO-INSTALLMENT INVESTIGATION")
print("=" * 65)

zero_installment_records = payments_clean[
    payments_clean["payment_installments"] == 0
].copy()

print("Records with zero installments:",
      len(zero_installment_records))

# Add order status to understand the context.
zero_installment_details = zero_installment_records.merge(
    orders_clean[["order_id", "order_status"]],
    on="order_id",
    how="left",
    validate="many_to_one"
)

print("\nZero-installment records with order status:")

print(
    zero_installment_details[
        [
            "order_id",
            "payment_type",
            "payment_value",
            "payment_installments",
            "order_status"
        ]
    ].to_string(index=False)
)


# ------------------------------------------------------------
# 3. INVESTIGATE ZERO-WEIGHT PRODUCTS
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("ZERO-WEIGHT PRODUCT PURCHASE ACTIVITY")
print("=" * 65)

# Count how often each product appears in order items.
zero_weight_ids = products_clean.loc[
    products_clean["product_weight_g"] == 0,
    "product_id"
]

zero_weight_activity = (
    order_items_clean[
        order_items_clean["product_id"].isin(zero_weight_ids)
    ]
    .groupby("product_id")
    .agg(
        item_records=("order_id", "count"),
        distinct_orders=("order_id", "nunique")
    )
    .reset_index()
)

zero_weight_details = products_clean[
    products_clean["product_weight_g"] == 0
].merge(
    zero_weight_activity,
    on="product_id",
    how="left",
    validate="one_to_one"
)

print(
    zero_weight_details[
        [
            "product_id",
            "product_category_name",
            "product_weight_g",
            "item_records",
            "distinct_orders"
        ]
    ].to_string(index=False)
)


# ------------------------------------------------------------
# 4. INVESTIGATE ORDER DATE SEQUENCES
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("ORDER DATE SEQUENCE VALIDATION")
print("=" * 65)

# Create flags without changing the original timestamps.
orders_clean["approval_before_purchase"] = (
    orders_clean["order_approved_at"]
    < orders_clean["order_purchase_timestamp"]
)

orders_clean["carrier_before_approval"] = (
    orders_clean["order_delivered_carrier_date"]
    < orders_clean["order_approved_at"]
)

orders_clean["delivery_before_carrier"] = (
    orders_clean["order_delivered_customer_date"]
    < orders_clean["order_delivered_carrier_date"]
)

orders_clean["delivery_after_estimate"] = (
    orders_clean["order_delivered_customer_date"]
    > orders_clean["order_estimated_delivery_date"]
)

date_flags = [
    "approval_before_purchase",
    "carrier_before_approval",
    "delivery_before_carrier",
    "delivery_after_estimate"
]

for flag in date_flags:
    print(f"{flag:30}: {orders_clean[flag].sum()}")


# ------------------------------------------------------------
# 5. DISPLAY RECORDS WITH IMPOSSIBLE DATE SEQUENCES
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("SAMPLE OF ORDER DATE INCONSISTENCIES")
print("=" * 65)

date_anomalies = orders_clean[
    orders_clean[
        [
            "approval_before_purchase",
            "carrier_before_approval",
            "delivery_before_carrier"
        ]
    ].any(axis=1)
].copy()

print("Orders with at least one date-sequence inconsistency:",
      len(date_anomalies))

print("\nSample records:")

print(
    date_anomalies[
        [
            "order_id",
            "order_status",
            "order_purchase_timestamp",
            "order_approved_at",
            "order_delivered_carrier_date",
            "order_delivered_customer_date",
            "approval_before_purchase",
            "carrier_before_approval",
            "delivery_before_carrier"
        ]
    ].head(15).to_string(index=False)
)


# ------------------------------------------------------------
# 6. FINAL INVESTIGATION SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("INVESTIGATION COMPLETED")
print("=" * 65)

print("No transactions or timestamps have been deleted or overwritten.")
print("Investigation flags have been added to orders_clean.")
```

```python

# ============================================================
# STEP 5: FINAL INVESTIGATION OF UNUSUAL ORDERS
# ============================================================

# ------------------------------------------------------------
# 1. COMPARE PAYMENT VALUES WITH ITEM VALUES
# ------------------------------------------------------------

# Aggregate item spending by order.
item_order_summary = (
    order_items_clean
    .groupby("order_id")
    .agg(
        item_spending=("price", "sum"),
        freight_spending=("freight_value", "sum"),
        item_count=("order_item_id", "count")
    )
    .reset_index()
)

# Combine payment, item and order information.
zero_payment_order_details = (
    zero_payment_orders
    .merge(
        item_order_summary,
        on="order_id",
        how="left",
        validate="one_to_one"
    )
    .merge(
        orders_clean[["order_id", "order_status"]],
        on="order_id",
        how="left",
        validate="one_to_one"
    )
)

print("=" * 65)
print("ZERO-PAYMENT ORDERS: BUSINESS CONTEXT")
print("=" * 65)

print(
    zero_payment_order_details[
        [
            "order_id",
            "order_status",
            "payment_record_count",
            "zero_payment_records",
            "positive_payment_records",
            "total_payment_value",
            "item_spending",
            "freight_spending",
            "item_count"
        ]
    ].to_string(index=False)
)


# ------------------------------------------------------------
# 2. CHECK ORDERS WITH DATE-SEQUENCE ANOMALIES BY STATUS
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("DATE ANOMALIES BY ORDER STATUS")
print("=" * 65)

for flag in [
    "carrier_before_approval",
    "delivery_before_carrier"
]:

    print(f"\n{flag.upper()}")

    print(
        orders_clean.groupby("order_status")[flag]
        .sum()
        .to_string()
    )


# ------------------------------------------------------------
# 3. MEASURE THE SIZE OF DATE-SEQUENCE DIFFERENCES
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("DATE-SEQUENCE TIME DIFFERENCES")
print("=" * 65)

# Calculate the number of hours between approval and carrier.
orders_clean["carrier_approval_gap_hours"] = (
    orders_clean["order_delivered_carrier_date"]
    - orders_clean["order_approved_at"]
).dt.total_seconds() / 3600

# Calculate the number of hours between carrier and delivery.
orders_clean["delivery_carrier_gap_hours"] = (
    orders_clean["order_delivered_customer_date"]
    - orders_clean["order_delivered_carrier_date"]
).dt.total_seconds() / 3600

print("\nCarrier timestamp earlier than approval:")

print(
    orders_clean.loc[
        orders_clean["carrier_before_approval"],
        "carrier_approval_gap_hours"
    ].describe().round(2)
)

print("\nDelivery timestamp earlier than carrier:")

print(
    orders_clean.loc[
        orders_clean["delivery_before_carrier"],
        "delivery_carrier_gap_hours"
    ].describe().round(2)
)


# ------------------------------------------------------------
# 4. CONFIRM DELIVERED ORDER ELIGIBILITY
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("ORDER STATUS AND SEGMENTATION ELIGIBILITY")
print("=" * 65)

print(
    orders_clean["order_status"]
    .value_counts(dropna=False)
    .to_string()
)

print("\nDelivered orders:",
      (orders_clean["order_status"] == "delivered").sum())

print("Non-delivered orders:",
      (orders_clean["order_status"] != "delivered").sum())


# ------------------------------------------------------------
# 5. CHECK ZERO-ONLY PAYMENT ORDERS THAT WERE DELIVERED
# ------------------------------------------------------------

zero_only_delivered = zero_payment_order_details[
    (zero_payment_order_details["positive_payment_records"] == 0)
    & (zero_payment_order_details["order_status"] == "delivered")
]

print("\nZero-only payment orders with delivered status:",
      len(zero_only_delivered))

print("\nInvestigation completed. No records were modified.")
```

```python

# ============================================================
# STAGE 4.1: TRANSFORM ORDER-ITEM DATA
# ============================================================

# 1. Join order items with product information
# This allows us to analyze product categories alongside purchases.

items_enriched = order_items_clean.merge(
    products_clean[["product_id", "product_category_name"]],
    on="product_id",
    how="left",
    validate="many_to_one"
)

# 2. Create an analytical category column.
# Missing categories are labeled "Unknown" for analysis.
# The original product_category_name column remains unchanged.

items_enriched["product_category_analytical"] = (
    items_enriched["product_category_name"].fillna("Unknown")
)

# 3. Aggregate item information for each order.
# Each order will have one summary record.

item_order_features = (
    items_enriched.groupby("order_id")
    .agg(
        item_spending=("price", "sum"),
        freight_spending=("freight_value", "sum"),
        total_items=("order_item_id", "count"),
        distinct_products=("product_id", "nunique"),
        category_diversity=(
            "product_category_analytical", "nunique"
        )
    )
    .reset_index()
)

# 4. Check the resulting table

print("ORDER-ITEM FEATURES")
print("-" * 45)
print("Shape:", item_order_features.shape)
print("Unique orders:", item_order_features["order_id"].nunique())
print("Duplicate order IDs:", item_order_features["order_id"].duplicated().sum())

display(item_order_features.head())
```

```python

# ============================================================
# STAGE 4.2: TRANSFORM PAYMENT DATA
# ============================================================

# 1. Define a function to identify the preferred payment method.
# Alphabetical order resolves ties consistently.

def preferred_payment_type(series):

    modes = series.mode(dropna=True)

    if len(modes) == 0:
        return "Unknown"

    return sorted(modes.astype(str))[0]


# 2. Aggregate payment information by order_id.
# Multiple payment records belonging to one order
# are combined into one summary record.

payment_order_features = (
    payments_clean.groupby("order_id")
    .agg(
        total_payment_value=("payment_value", "sum"),
        payment_record_count=("payment_value", "count"),
        preferred_payment_type=(
            "payment_type", preferred_payment_type
        )
    )
    .reset_index()
)


# 3. Validate the payment feature table

print("PAYMENT FEATURES")
print("-" * 45)

print("Shape:", payment_order_features.shape)
print(
    "Unique orders:",
    payment_order_features["order_id"].nunique()
)
print(
    "Duplicate order IDs:",
    payment_order_features["order_id"].duplicated().sum()
)

display(payment_order_features.head())

```

```python

# ============================================================
# STAGE 4.3: TRANSFORM REVIEW DATA
# ============================================================

# Aggregate review information for each order.
# An order may have more than one review record.

review_order_features = (
    reviews_clean.groupby("order_id")
    .agg(
        average_review_score=("review_score", "mean"),
        review_record_count=("review_score", "count")
    )
    .reset_index()
)

# Validate the review feature table

print("REVIEW FEATURES")
print("-" * 45)

print("Shape:", review_order_features.shape)
print(
    "Unique orders:",
    review_order_features["order_id"].nunique()
)
print(
    "Duplicate order IDs:",
    review_order_features["order_id"].duplicated().sum()
)

display(review_order_features.head())
```

```python
# ============================================================
# STAGE 4.4': BUILD THE COMPLETE UNIFIED ORDER-LEVEL DATASET
# ============================================================

# 1. Prepare customer lookup
customer_lookup = customers_clean[
    [
        "customer_id",
        "customer_unique_id",
        "customer_zip_code_prefix",
        "customer_city",
        "customer_state"
    ]
].copy()


# 2. Start with ALL orders, not only delivered orders
order_level_all = orders_clean.copy()


# 3. Integrate customer information
order_level_all = order_level_all.merge(
    customer_lookup,
    on="customer_id",
    how="left",
    validate="many_to_one"
)


# 4. Integrate aggregated order-item information
order_level_all = order_level_all.merge(
    item_order_features,
    on="order_id",
    how="left",
    validate="one_to_one"
)


# 5. Integrate aggregated payment information
order_level_all = order_level_all.merge(
    payment_order_features,
    on="order_id",
    how="left",
    validate="one_to_one"
)


# 6. Integrate aggregated review information
order_level_all = order_level_all.merge(
    review_order_features,
    on="order_id",
    how="left",
    validate="one_to_one"
)


# 7. Create the delivered-order analytical dataset
order_level = order_level_all[
    order_level_all["order_status"] == "delivered"
].copy()


# ============================================================
# VALIDATION
# ============================================================

print("=" * 60)
print("UNIFIED ORDER-LEVEL DATASET")
print("=" * 60)

print("Unified dataset shape:", order_level_all.shape)
print("Unique orders:", order_level_all["order_id"].nunique())
print("Duplicate order IDs:", order_level_all["order_id"].duplicated().sum())

print("\nDelivered-order analytical dataset")
print("Shape:", order_level.shape)
print("Unique delivered orders:", order_level["order_id"].nunique())

print("\nMissing values in unified dataset:")
print(
    order_level_all[
        [
            "customer_unique_id",
            "item_spending",
            "total_payment_value",
            "average_review_score"
        ]
    ].isna().sum()
)

print("\nOrder status distribution:")
print(order_level_all["order_status"].value_counts())

print("\nSample unified records:")
display(order_level_all.head())
```

```python

# ============================================================
# STAGE 6.1: RAW DATA INGESTION AND CLEANING
# ============================================================

import pandas as pd
from pathlib import Path
from prefect import task

# Location of the original Olist CSV files
RAW_DATA_DIR = Path(
    "C:/Users/HP/Documents/Nexford_Project_File"
)


@task(name="Ingest and Clean Raw Data")
def ingest_and_clean_raw_data():

    # 1. Define the six raw data sources
    data_sources = {
        "customers": "olist_customers_dataset.csv",
        "orders": "olist_orders_dataset.csv",
        "order_items": "olist_order_items_dataset.csv",
        "payments": "olist_order_payments_dataset.csv",
        "reviews": "olist_order_reviews_dataset.csv",
        "products": "olist_products_dataset.csv"
    }

    cleaned_raw = {}

    # 2. Load and standardize each dataset
    for dataset_name, filename in data_sources.items():

        file_path = RAW_DATA_DIR / filename

        # Check that the source file exists
        if not file_path.exists():
            raise FileNotFoundError(
                f"Missing raw dataset: {file_path}"
            )

        # Load the raw CSV file
        df = pd.read_csv(
            file_path,
            encoding="utf-8",
            low_memory=False
        )

        # Work on a copy
        df = df.copy()

        # 3. Standardise identifier columns
        identifier_columns = {
            "customers": [
                "customer_id",
                "customer_unique_id"
            ],
            "orders": [
                "order_id",
                "customer_id"
            ],
            "order_items": [
                "order_id",
                "product_id",
                "seller_id"
            ],
            "payments": [
                "order_id"
            ],
            "reviews": [
                "review_id",
                "order_id"
            ],
            "products": [
                "product_id"
            ]
        }

        for column in identifier_columns[dataset_name]:
            df[column] = df[column].astype("string")

        # 4. Standardize customer ZIP prefixes
        if dataset_name == "customers":
            df["customer_zip_code_prefix"] = (
                df["customer_zip_code_prefix"]
                .astype("string")
                .str.zfill(5)
            )

        # 5. Convert date columns into datetime format
        date_columns = {
            "orders": [
                "order_purchase_timestamp",
                "order_approved_at",
                "order_delivered_carrier_date",
                "order_delivered_customer_date",
                "order_estimated_delivery_date"
            ],
            "reviews": [
                "review_creation_date",
                "review_answer_timestamp"
            ]
        }

        for column in date_columns.get(dataset_name, []):
            df[column] = pd.to_datetime(
                df[column],
                errors="coerce"
            )

        # 6. Store the cleaned dataset
        cleaned_raw[dataset_name] = df

        print(
            f"{dataset_name}: {df.shape} | "
            f"Standardisation completed"
        )

    print("\nRAW DATA INGESTION AND CLEANING COMPLETED.")

    return cleaned_raw


# 7. Execute the task once to test it
cleaned_raw_datasets = ingest_and_clean_raw_data()
```

```python

# ============================================================
# STAGE 6.2: AGGREGATE TRANSACTION DATA
# ============================================================

from prefect import task, flow


@task(name="Aggregate Order Transactions")
def aggregate_order_data(cleaned_raw):

    # Retrieve the cleaned raw datasets
    items = cleaned_raw["order_items"].copy()
    products = cleaned_raw["products"].copy()
    payments = cleaned_raw["payments"].copy()
    reviews = cleaned_raw["reviews"].copy()

    # --------------------------------------------------------
    # 1. ITEM AGGREGATION
    # --------------------------------------------------------

    # Attach product categories to order items
    items_enriched = items.merge(
        products[["product_id", "product_category_name"]],
        on="product_id",
        how="left",
        validate="many_to_one"
    )

    # Preserve original missing categories while using
    # "Unknown" for analytical category counting
    items_enriched["product_category_analytical"] = (
        items_enriched["product_category_name"]
        .fillna("Unknown")
    )

    # Summarize item transactions by order
    item_features = (
        items_enriched.groupby("order_id")
        .agg(
            item_spending=("price", "sum"),
            freight_spending=("freight_value", "sum"),
            total_items=("order_item_id", "count"),
            distinct_products=("product_id", "nunique"),
            category_diversity=(
                "product_category_analytical", "nunique"
            )
        )
        .reset_index()
    )

    # --------------------------------------------------------
    # 2. PAYMENT AGGREGATION
    # --------------------------------------------------------

    # Select the most frequent payment type.
    # Alphabetical ordering resolves ties consistently.
    def preferred_payment_type(series):
        counts = series.value_counts()
        highest_count = counts.max()
        tied_types = sorted(
            counts[counts == highest_count].index.tolist()
        )
        return tied_types[0]

    payment_features = (
        payments.groupby("order_id")
        .agg(
            total_payment_value=("payment_value", "sum"),
            payment_record_count=("payment_value", "count"),
            preferred_payment_type=(
                "payment_type", preferred_payment_type
            )
        )
        .reset_index()
    )

    # --------------------------------------------------------
    # 3. REVIEW AGGREGATION
    # --------------------------------------------------------

    # Group by order_id rather than review_id because
    # review IDs are reused across some orders.
    review_features = (
        reviews.groupby("order_id")
        .agg(
            average_review_score=("review_score", "mean"),
            review_record_count=("review_score", "count")
        )
        .reset_index()
    )

    # --------------------------------------------------------
    # 4. BASIC INTEGRITY CHECKS
    # --------------------------------------------------------

    summaries = {
        "item_features": item_features,
        "payment_features": payment_features,
        "review_features": review_features
    }

    for name, dataframe in summaries.items():

        if dataframe["order_id"].duplicated().any():
            raise ValueError(
                f"Duplicate order IDs found in {name}."
            )

        print(
            f"{name}: {dataframe.shape} | "
            "Unique order IDs confirmed"
        )

    print("\nTRANSACTION AGGREGATION COMPLETED.")

    return summaries


# Execute the task inside a Prefect flow
@flow(name="Test Olist Transaction Aggregation")
def test_transaction_aggregation():

    return aggregate_order_data(cleaned_raw_datasets)


aggregated_data = test_transaction_aggregation()

# Keep the three summaries available for the next stage
item_order_features = aggregated_data["item_features"]
payment_order_features = aggregated_data["payment_features"]
review_order_features = aggregated_data["review_features"]
```

```python

# ============================================================
# STAGE 6.3: INTEGRATE ORDER AND CUSTOMER DATA
# ============================================================

from prefect import flow, task


@task(name="Integrate Order and Customer Data")
def integrate_order_data(cleaned_raw, aggregated_data):

    # Retrieve cleaned source datasets
    orders = cleaned_raw["orders"].copy()
    customers = cleaned_raw["customers"].copy()

    # Retrieve aggregated transaction datasets
    item_features = aggregated_data["item_features"]
    payment_features = aggregated_data["payment_features"]
    review_features = aggregated_data["review_features"]

    # --------------------------------------------------------
    # 1. Attach customer information to orders
    # --------------------------------------------------------

    customer_lookup = customers[
        [
            "customer_id",
            "customer_unique_id",
            "customer_zip_code_prefix",
            "customer_city",
            "customer_state"
        ]
    ].copy()

    unified_orders = orders.merge(
        customer_lookup,
        on="customer_id",
        how="left",
        validate="many_to_one"
    )

    # --------------------------------------------------------
    # 2. Integrate item, payment and review summaries
    # --------------------------------------------------------

    unified_orders = unified_orders.merge(
        item_features,
        on="order_id",
        how="left",
        validate="one_to_one"
    )

    unified_orders = unified_orders.merge(
        payment_features,
        on="order_id",
        how="left",
        validate="one_to_one"
    )

    unified_orders = unified_orders.merge(
        review_features,
        on="order_id",
        how="left",
        validate="one_to_one"
    )

    # --------------------------------------------------------
    # 3. Create missing-date indicators
    # --------------------------------------------------------

    unified_orders["approval_date_missing"] = (
        unified_orders["order_approved_at"].isna()
    )

    unified_orders["carrier_date_missing"] = (
        unified_orders["order_delivered_carrier_date"].isna()
    )

    unified_orders["delivery_date_missing"] = (
        unified_orders["order_delivered_customer_date"].isna()
    )

    # --------------------------------------------------------
    # 4. Create date-quality flags
    # --------------------------------------------------------

    unified_orders["approval_before_purchase"] = (
        unified_orders["order_approved_at"]
        < unified_orders["order_purchase_timestamp"]
    )

    unified_orders["carrier_before_approval"] = (
        unified_orders["order_delivered_carrier_date"]
        < unified_orders["order_approved_at"]
    )

    unified_orders["delivery_before_carrier"] = (
        unified_orders["order_delivered_customer_date"]
        < unified_orders["order_delivered_carrier_date"]
    )

    unified_orders["delivery_after_estimate"] = (
        unified_orders["order_delivered_customer_date"]
        > unified_orders["order_estimated_delivery_date"]
    )

    # Calculate time gaps without changing original timestamps
    unified_orders["carrier_approval_gap_hours"] = (
        unified_orders["order_delivered_carrier_date"]
        - unified_orders["order_approved_at"]
    ).dt.total_seconds() / 3600

    unified_orders["delivery_carrier_gap_hours"] = (
        unified_orders["order_delivered_customer_date"]
        - unified_orders["order_delivered_carrier_date"]
    ).dt.total_seconds() / 3600

    # --------------------------------------------------------
    # 5. Validate the integrated dataset
    # --------------------------------------------------------

    if unified_orders["order_id"].duplicated().any():
        raise ValueError("Duplicate order IDs found.")

    if unified_orders["customer_unique_id"].isna().any():
        raise ValueError("Some orders have no customer identity.")

    # Create delivered-order dataset
    delivered_orders = unified_orders[
        unified_orders["order_status"] == "delivered"
    ].copy()

    print("INTEGRATION COMPLETED")
    print("-" * 45)
    print("Unified dataset:", unified_orders.shape)
    print("Delivered dataset:", delivered_orders.shape)
    print(
        "Unique unified orders:",
        unified_orders["order_id"].nunique()
    )
    print(
        "Unique delivered orders:",
        delivered_orders["order_id"].nunique()
    )

    print("\nDelivered order status check:")
    print(delivered_orders["order_status"].value_counts())

    return unified_orders, delivered_orders


# Execute integration within a Prefect flow
@flow(name="Test Olist Data Integration")
def test_data_integration():

    return integrate_order_data(
        cleaned_raw_datasets,
        aggregated_data
    )


unified_orders, delivered_orders = test_data_integration()

```

```python

# ============================================================
# STAGE 6.5: COMPLETE AUTOMATED OLIST DATA PIPELINE
# ============================================================

import great_expectations as gx
import pandas as pd
from pathlib import Path
from prefect import flow, task

PROJECT_DIR = Path.cwd()
PROCESSED_DIR = PROJECT_DIR / "data" / "processed"
GX_PROJECT_DIR = PROJECT_DIR / "gx"


# ------------------------------------------------------------
# 1. VALIDATE THE SAVED UNIFIED DATASET
# ------------------------------------------------------------

@task(name="Validate Final Unified Dataset")
def validate_final_dataset():

    # Load the saved Great Expectations context
    gx_context = gx.get_context(
        context_root_dir=str(GX_PROJECT_DIR)
    )

    # Retrieve the clean nine-expectation validation definition
    validation_definition = (
        gx_context.validation_definitions.get(
            "olist_order_quality_validation_v2"
        )
    )

    # Load the newly generated unified dataset
    unified_data = pd.read_csv(
        PROCESSED_DIR / "unified_order_level.csv"
    )

    # Execute validation
    validation = validation_definition.run(
        batch_parameters={"dataframe": unified_data}
    )

    print("\nFINAL DATA QUALITY VALIDATION")
    print("-" * 45)
    print("Validation successful:", validation.success)
    print(
        "Total checks:",
        validation.statistics["evaluated_expectations"]
    )
    print(
        "Successful checks:",
        validation.statistics["successful_expectations"]
    )
    print(
        "Failed checks:",
        validation.statistics["unsuccessful_expectations"]
    )

    # Stop the pipeline if validation fails
    if not validation.success:
        raise ValueError(
            "Final data quality validation failed."
        )

    return True


# ------------------------------------------------------------
# 2. DEFINE THE COMPLETE PIPELINE
# ------------------------------------------------------------

@flow(name="Olist End-to-End Data Pipeline")
def olist_end_to_end_pipeline():

    print("=" * 60)
    print("STARTING OLIST END-TO-END DATA PIPELINE")
    print("=" * 60)

    # Stage A: Load and standardize raw datasets
    cleaned_raw = ingest_and_clean_raw_data()

    # Stage B: Aggregate item, payment and review records
    aggregated = aggregate_order_data(cleaned_raw)

    # Stage C: Integrate orders and customer information
    unified, delivered = integrate_order_data(
        cleaned_raw,
        aggregated
    )

    # Stage D: Save the integrated datasets
    saved_paths = save_integrated_data(
        unified,
        delivered
    )

    # Stage E: Validate the final unified dataset
    validation_passed = validate_final_dataset()

    # Final pipeline summary
    print("\n" + "=" * 60)
    print("OLIST PIPELINE EXECUTION SUMMARY")
    print("=" * 60)

    print("Raw ingestion and standardisation: SUCCESS")
    print("Transaction aggregation: SUCCESS")
    print("Data integration: SUCCESS")
    print("Dataset saving: SUCCESS")
    print("Great Expectations validation:",
          "SUCCESS" if validation_passed else "FAILED")

    print("\nUnified dataset:", unified.shape)
    print("Delivered dataset:", delivered.shape)

    print("\nOVERALL PIPELINE: SUCCESS")

    return {
        "unified_shape": unified.shape,
        "delivered_shape": delivered.shape,
        "saved_paths": saved_paths,
        "validation_passed": validation_passed
    }


# ------------------------------------------------------------
# 3. EXECUTE THE COMPLETE PIPELINE
# ------------------------------------------------------------

pipeline_results = olist_end_to_end_pipeline()
```

```python

# ============================================================
# STAGE 7.1: CUSTOMER REPRESENTATION BIAS ASSESSMENT
# ============================================================

import pandas as pd
from pathlib import Path
from prefect import flow, task


@task(name="Assess Geographic Representation")
def assess_geographic_representation(delivered_data):

    # Count unique customers represented in each state
    state_summary = (
        delivered_data.groupby("customer_state")
        .agg(
            unique_customers=("customer_unique_id", "nunique"),
            delivered_orders=("order_id", "nunique")
        )
        .reset_index()
    )

    # Calculate each state's share of the customer population
    total_customers = state_summary["unique_customers"].sum()

    state_summary["customer_percentage"] = (
        state_summary["unique_customers"]
        / total_customers * 100
    )

    # Sort states by customer representation
    state_summary = state_summary.sort_values(
        "unique_customers",
        ascending=False
    ).reset_index(drop=True)

    # Flag groups with fewer than 100 unique customers
    state_summary["small_sample_flag"] = (
        state_summary["unique_customers"] < 100
    )

    # Display results
    print("GEOGRAPHIC REPRESENTATION ASSESSMENT")
    print("-" * 55)
    print("Number of states:", len(state_summary))
    print("Total unique customers:", total_customers)

    print("\nTop 10 states by customer representation:")
    print(state_summary.head(10).to_string(index=False))

    print("\nStates with fewer than 100 customers:")
    print(
        state_summary[
            state_summary["small_sample_flag"]
        ].to_string(index=False)
    )

    print("\nRepresentation summary:")
    print(
        "Largest state share:",
        round(
            state_summary["customer_percentage"].max(), 2
        ),
        "%"
    )

    print(
        "Small-sample states:",
        state_summary["small_sample_flag"].sum()
    )

    # Save the assessment results
    output_dir = Path.cwd() / "reports" / "governance"
    output_dir.mkdir(parents=True, exist_ok=True)

    output_path = (
        output_dir / "geographic_representation_audit.csv"
    )

    state_summary.to_csv(output_path, index=False)

    print("\nAssessment saved to:", output_path)

    return state_summary


@flow(name="Olist Geographic Representation Audit")
def run_representation_audit():

    return assess_geographic_representation(
        delivered_orders
    )


representation_results = run_representation_audit()
```

```python

import ast
import json
import os
from pathlib import Path

# ------------------------------------------------------------
# 1. LOAD THE EXISTING NOTEBOOK
# ------------------------------------------------------------

project_dir = Path.cwd()
notebook_path = project_dir / "data_cleaning.ipynb"

with open(notebook_path, "r", encoding="utf-8") as file:
    notebook = json.load(file)

# Functions required for the complete pipeline
required_functions = {
    "ingest_and_clean_raw_data",
    "aggregate_order_data",
    "integrate_order_data",
    "save_integrated_data",
    "validate_final_dataset",
    "olist_end_to_end_pipeline"
}

extracted_functions = {}

# ------------------------------------------------------------
# 2. EXTRACT THE ORIGINAL FUNCTION DEFINITIONS
# ------------------------------------------------------------

for cell_index in range(28, 33):

    cell = notebook["cells"][cell_index]

    if cell["cell_type"] != "code":
        continue

    source = "".join(cell["source"])
    syntax_tree = ast.parse(source)

    lines = source.splitlines()

    for node in syntax_tree.body:

        if not isinstance(node, (ast.FunctionDef, ast.AsyncFunctionDef)):
            continue

        if node.name not in required_functions:
            continue

        # Include decorators such as @task and @flow
        start_line = min(
            [node.lineno] +
            [decorator.lineno for decorator in node.decorator_list]
        )

        end_line = node.end_lineno

        function_source = "\n".join(
            lines[start_line - 1:end_line]
        )

        extracted_functions[node.name] = function_source


# Confirm all required functions were found
missing_functions = required_functions - set(extracted_functions)

if missing_functions:
    raise ValueError(
        f"Could not locate these functions: {missing_functions}"
    )


# ------------------------------------------------------------
# 3. CREATE PORTABLE PROJECT CONFIGURATION
# ------------------------------------------------------------

header = r'''
# ============================================================
# OLIST END-TO-END DATA PIPELINE
# Extracted from the tested project notebook
# ============================================================

import os
from pathlib import Path

import pandas as pd
import great_expectations as gx

from prefect import flow, task

# Project directory: location of this Python script
PROJECT_DIR = Path(__file__).resolve().parent

# Raw data directory can be configured through an environment
# variable. Docker will mount the source data into /app/data/raw.
RAW_DATA_DIR = Path(
    os.environ.get(
        "OLIST_RAW_DATA_DIR",
        str(PROJECT_DIR / "data" / "raw")
    )
)

# Processed datasets and Great Expectations configuration
PROCESSED_DIR = PROJECT_DIR / "data" / "processed"
GX_PROJECT_DIR = PROJECT_DIR / "gx"

'''


# ------------------------------------------------------------
# 4. ASSEMBLE THE STANDALONE PIPELINE
# ------------------------------------------------------------

function_order = [
    "ingest_and_clean_raw_data",
    "aggregate_order_data",
    "integrate_order_data",
    "save_integrated_data",
    "validate_final_dataset",
    "olist_end_to_end_pipeline"
]

script_parts = [header]

for function_name in function_order:

    script_parts.append(
        extracted_functions[function_name]
    )

    script_parts.append("\n\n")


# Execute the complete pipeline when the script is run
script_parts.append(
    '''
if __name__ == "__main__":

    pipeline_results = olist_end_to_end_pipeline()

    print("\\nPIPELINE RESULTS")
    print(pipeline_results)
    '''
)

# ------------------------------------------------------------
# 5. SAVE THE PYTHON SCRIPT
# ------------------------------------------------------------

script_path = project_dir / "olist_pipeline.py"

script_path.write_text(
    "\n".join(script_parts),
    encoding="utf-8"
)

print("PIPELINE SCRIPT CREATED SUCCESSFULLY")
print("=" * 55)
print("File:", script_path)
print("\nExtracted functions:")

for function_name in function_order:
    print("-", function_name)

print("\nScript size:", script_path.stat().st_size, "bytes")
```
