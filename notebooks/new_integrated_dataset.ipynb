```python

# ============================================================
#  IMPORT LIBRARIES AND LOAD THE OLIST DATASETS
# Project: Low Customer Satisfaction Review Prediction
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path

# Display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
pd.set_option("display.width", 120)

# Locate the project directory.
# This assumes the notebook is saved inside the notebooks folder.
PROJECT_ROOT = Path.cwd().parent

# Folder containing the nine original CSV files
RAW_DATA_PATH = PROJECT_ROOT / "data" / "raw"

# Define the nine original Olist datasets
file_names = {
    "customers": "olist_customers_dataset.csv",
    "orders": "olist_orders_dataset.csv",
    "order_items": "olist_order_items_dataset.csv",
    "payments": "olist_order_payments_dataset.csv",
    "reviews": "olist_order_reviews_dataset.csv",
    "products": "olist_products_dataset.csv",
    "sellers": "olist_sellers_dataset.csv",
    "geolocation": "olist_geolocation_dataset.csv",
    "category_translation": "product_category_name_translation.csv"
}

# Load each CSV into a separate DataFrame
datasets = {}

for name, file_name in file_names.items():

    file_path = RAW_DATA_PATH / file_name

    if not file_path.exists():
        print(f"FILE NOT FOUND: {file_name}")
        continue

    datasets[name] = pd.read_csv(file_path)

    print(f"Loaded: {file_name}")
    print(f"Rows: {datasets[name].shape[0]:,}")
    print(f"Columns: {datasets[name].shape[1]}")
    print("-" * 60)

# Confirm loading results
print("\nDATASET LOADING SUMMARY")
print("=" * 60)

print(f"Expected datasets: {len(file_names)}")
print(f"Successfully loaded: {len(datasets)}")

if len(datasets) == len(file_names):
    print("All nine Olist datasets loaded successfully.")
else:
    print("Some datasets are missing. Check the filenames and folder.")
```

```python

# ============================================================
#  INSPECT DATASET STRUCTURE
# ============================================================

for name, df in datasets.items():

    print("\n" + "=" * 80)
    print(f"DATASET: {name.upper()}")
    print("=" * 80)

    print(f"\nShape: {df.shape}")

    print("\nColumn names:")
    print(df.columns.tolist())

    print("\nData types:")
    print(df.dtypes)

    print("\nFirst five records:")
    display(df.head())
```

```python

# ============================================================
# DATA QUALITY ASSESSMENT
# ============================================================

# ------------------------------------------------------------
# Missing values
# ------------------------------------------------------------

print("MISSING VALUE SUMMARY")
print("=" * 80)

for name, df in datasets.items():

    missing = df.isnull().sum()
    missing = missing[missing > 0]

    print(f"\n{name.upper()}")

    if missing.empty:
        print("No missing values found.")
    else:
        missing_summary = pd.DataFrame({
            "Missing Count": missing,
            "Missing Percentage": (missing / len(df) * 100).round(2)
        })

        display(missing_summary)


# ------------------------------------------------------------
#  Duplicate records
# ------------------------------------------------------------

print("\n\nDUPLICATE RECORD SUMMARY")
print("=" * 80)

for name, df in datasets.items():

    print(
        f"{name}: "
        f"{df.duplicated().sum():,} completely duplicated rows"
    )


# ------------------------------------------------------------
#  Check important identifier uniqueness
# ------------------------------------------------------------

print("\n\nIDENTIFIER UNIQUENESS CHECK")
print("=" * 80)

unique_checks = {
    "customers": ["customer_id", "customer_unique_id"],
    "orders": ["order_id"],
    "order_items": ["order_id", "order_item_id"],
    "payments": ["order_id", "payment_sequential"],
    "reviews": ["review_id"],
    "products": ["product_id"],
    "sellers": ["seller_id"],
    "category_translation": ["product_category_name"]
}

for dataset_name, columns in unique_checks.items():

    df = datasets[dataset_name]

    print(f"\n{dataset_name.upper()}")

    for column in columns:

        duplicate_count = df[column].duplicated().sum()

        print(
            f"{column}: "
            f"{duplicate_count:,} duplicate values"
        )


# ------------------------------------------------------------
#  Examine order status distribution
# ------------------------------------------------------------

print("\n\nORDER STATUS DISTRIBUTION")
print("=" * 80)

display(
    datasets["orders"]["order_status"]
    .value_counts(dropna=False)
    .to_frame("Number of Orders")
)


# ------------------------------------------------------------
#  Examine review score distribution
# ------------------------------------------------------------

print("\n\nREVIEW SCORE DISTRIBUTION")
print("=" * 80)

review_distribution = (
    datasets["reviews"]["review_score"]
    .value_counts(dropna=False)
    .sort_index()
    .to_frame("Number of Reviews")
)

review_distribution["Percentage"] = (
    review_distribution["Number of Reviews"]
    / len(datasets["reviews"]) * 100
).round(2)

display(review_distribution)


# ------------------------------------------------------------
#  Check number of records per order
# ------------------------------------------------------------

print("\n\nRECORDS PER ORDER")
print("=" * 80)

for name in ["order_items", "payments", "reviews"]:

    df = datasets[name]

    records_per_order = df.groupby("order_id").size()

    print(f"\n{name.upper()}")
    print(f"Orders represented: {records_per_order.size:,}")
    print(f"Maximum records per order: {records_per_order.max()}")
    print(f"Orders with multiple records: "
          f"{(records_per_order > 1).sum():,}")


# ------------------------------------------------------------
#  Check date columns
# ------------------------------------------------------------

print("\n\nDATE COLUMN INSPECTION")
print("=" * 80)

date_columns = {
    "orders": [
        "order_purchase_timestamp",
        "order_approved_at",
        "order_delivered_carrier_date",
        "order_delivered_customer_date",
        "order_estimated_delivery_date"
    ],
    "order_items": ["shipping_limit_date"],
    "reviews": [
        "review_creation_date",
        "review_answer_timestamp"
    ]
}

for dataset_name, columns in date_columns.items():

    print(f"\n{dataset_name.upper()}")

    for column in columns:

        converted = pd.to_datetime(
            datasets[dataset_name][column],
            errors="coerce"
        )

        print(
            f"{column}: "
            f"{converted.isna().sum():,} missing or invalid dates"
        )
```

```python

# ============================================================
#  INVESTIGATE DUPLICATES AND TARGET RELIABILITY
# ============================================================

orders = datasets["orders"]
reviews = datasets["reviews"]
items = datasets["order_items"]
payments = datasets["payments"]
geolocation = datasets["geolocation"]


# ------------------------------------------------------------
# Investigate duplicate review IDs
# ------------------------------------------------------------

print("DUPLICATE REVIEW ID INVESTIGATION")
print("=" * 70)

duplicate_review_ids = reviews[
    reviews["review_id"].duplicated(keep=False)
].sort_values("review_id")

print("Rows with repeated review IDs:",
      len(duplicate_review_ids))

display(duplicate_review_ids.head(20))

print("\nNumber of distinct repeated review IDs:",
      duplicate_review_ids["review_id"].nunique())


# ------------------------------------------------------------
#  Investigate orders with multiple reviews
# ------------------------------------------------------------

print("\n\nMULTIPLE REVIEWS PER ORDER")
print("=" * 70)

review_counts = reviews.groupby("order_id").size()

multiple_review_orders = review_counts[review_counts > 1]

print("Orders with multiple reviews:",
      len(multiple_review_orders))

multiple_review_details = (
    reviews[
        reviews["order_id"].isin(multiple_review_orders.index)
    ]
    .sort_values(["order_id", "review_creation_date"])
)

display(multiple_review_details.head(30))


# Check whether multiple reviews have conflicting scores
review_score_counts = (
    reviews.groupby("order_id")["review_score"].nunique()
)

conflicting_review_orders = review_score_counts[
    review_score_counts > 1
]

print("\nOrders with conflicting review scores:",
      len(conflicting_review_orders))


# ------------------------------------------------------------
# 4.3 Investigate review coverage for delivered orders
# ------------------------------------------------------------

print("\n\nDELIVERED ORDER REVIEW COVERAGE")
print("=" * 70)

delivered_orders = orders[
    orders["order_status"] == "delivered"
]

reviewed_delivered = delivered_orders[
    delivered_orders["order_id"].isin(reviews["order_id"])
]

unreviewed_delivered = delivered_orders[
    ~delivered_orders["order_id"].isin(reviews["order_id"])
]

print("Delivered orders:", len(delivered_orders))
print("Delivered orders with reviews:", len(reviewed_delivered))
print("Delivered orders without reviews:", len(unreviewed_delivered))


# ------------------------------------------------------------
#  Missing delivery dates by order status
# ------------------------------------------------------------

print("\n\nDELIVERY DATE MISSINGNESS BY ORDER STATUS")
print("=" * 70)

delivery_missingness = (
    orders.groupby("order_status")
    .agg(
        total_orders=("order_id", "count"),
        missing_carrier_date=(
            "order_delivered_carrier_date",
            lambda x: x.isna().sum()
        ),
        missing_customer_date=(
            "order_delivered_customer_date",
            lambda x: x.isna().sum()
        )
    )
)

display(delivery_missingness)


# ------------------------------------------------------------
#  Check review timing relative to delivery
# ------------------------------------------------------------

print("\n\nREVIEW TIMING CHECK")
print("=" * 70)

timing_data = reviews[
    ["order_id", "review_creation_date", "review_score"]
].merge(
    orders[
        ["order_id", "order_status",
         "order_delivered_customer_date"]
    ],
    on="order_id",
    how="left",
    validate="many_to_one"
)

timing_data["review_creation_date"] = pd.to_datetime(
    timing_data["review_creation_date"],
    errors="coerce"
)

timing_data["order_delivered_customer_date"] = pd.to_datetime(
    timing_data["order_delivered_customer_date"],
    errors="coerce"
)

timing_data["days_delivery_to_review"] = (
    timing_data["review_creation_date"]
    - timing_data["order_delivered_customer_date"]
).dt.total_seconds() / 86400

delivered_review_timing = timing_data[
    timing_data["order_status"] == "delivered"
]

print("Reviews linked to delivered orders:",
      len(delivered_review_timing))

print("\nDelivery-to-review interval in days:")
display(
    delivered_review_timing[
        "days_delivery_to_review"
    ].describe()
)

print("\nReviews created before recorded delivery:")
print(
    (delivered_review_timing["days_delivery_to_review"] < 0).sum()
)


# ------------------------------------------------------------
#  Investigate geolocation duplicates
# ------------------------------------------------------------

print("\n\nGEOLOCATION DUPLICATE INVESTIGATION")
print("=" * 70)

print("Total geolocation rows:", len(geolocation))

print("Unique ZIP-code prefixes:",
      geolocation["geolocation_zip_code_prefix"].nunique())

print("Unique ZIP prefixes and coordinate combinations:",
      geolocation[
          ["geolocation_zip_code_prefix",
           "geolocation_lat",
           "geolocation_lng"]
      ].drop_duplicates().shape[0])

print("\nRepeated ZIP-code prefixes:")
zip_counts = geolocation.groupby(
    "geolocation_zip_code_prefix"
).size()

print("ZIP prefixes with multiple records:",
      (zip_counts > 1).sum())

print("\nExample repeated ZIP prefixes:")
display(
    geolocation[
        geolocation["geolocation_zip_code_prefix"].isin(
            zip_counts[zip_counts > 1].head(5).index
        )
    ].head(20)
)


# ------------------------------------------------------------
# Validate order-item composite identifier
# ------------------------------------------------------------

print("\n\nORDER ITEM KEY VALIDATION")
print("=" * 70)

print(
    "Duplicate order_id + order_item_id combinations:",
    items.duplicated(
        subset=["order_id", "order_item_id"]
    ).sum()
)


# ------------------------------------------------------------
#  Validate payment composite identifier
# ------------------------------------------------------------

print("\n\nPAYMENT KEY VALIDATION")
print("=" * 70)

print(
    "Duplicate order_id + payment_sequential combinations:",
    payments.duplicated(
        subset=["order_id", "payment_sequential"]
    ).sum()
)
```

```python

# ============================================================
#  DETAILED REVIEW DUPLICATE INVESTIGATION
# ============================================================

import pandas as pd

reviews = datasets["reviews"].copy()

# ------------------------------------------------------------
# 1. Check whether repeated review IDs are linked to
#    multiple orders or contain different review information.
# ------------------------------------------------------------

review_id_summary = (
    reviews.groupby("review_id")
    .agg(
        row_count=("review_id", "size"),
        unique_orders=("order_id", "nunique"),
        unique_scores=("review_score", "nunique"),
        unique_creation_dates=("review_creation_date", "nunique"),
        unique_comments=("review_comment_message", "nunique")
    )
)

repeated_id_summary = review_id_summary[
    review_id_summary["row_count"] > 1
]

print("REPEATED REVIEW ID SUMMARY")
print("=" * 65)
print("Repeated review IDs:", len(repeated_id_summary))
print(
    "Repeated IDs linked to multiple orders:",
    (repeated_id_summary["unique_orders"] > 1).sum()
)
print(
    "Repeated IDs with different scores:",
    (repeated_id_summary["unique_scores"] > 1).sum()
)
print(
    "Repeated IDs with different creation dates:",
    (repeated_id_summary["unique_creation_dates"] > 1).sum()
)

# Display examples of repeated IDs
repeated_ids = repeated_id_summary.index

print("\nEXAMPLES OF REPEATED REVIEW IDs")
display(
    reviews[reviews["review_id"].isin(repeated_ids)]
    .sort_values(["review_id", "order_id"])
    .head(20)
)

# ------------------------------------------------------------
# 2. Check exact duplicate review records.
#    Exclude review_id and order_id when comparing the
#    review content and timestamps.
# ------------------------------------------------------------

review_content_columns = [
    "review_score",
    "review_comment_title",
    "review_comment_message",
    "review_creation_date",
    "review_answer_timestamp"
]

exact_content_duplicates = reviews.duplicated(
    subset=["order_id"] + review_content_columns,
    keep=False
)

print("\nDUPLICATE REVIEW CONTENT WITHIN THE SAME ORDER")
print("=" * 65)
print(
    "Rows involved in duplicate review content:",
    exact_content_duplicates.sum()
)

# ------------------------------------------------------------
# 3. Examine orders with conflicting review scores.
# ------------------------------------------------------------

score_counts = (
    reviews.groupby("order_id")["review_score"]
    .nunique()
)

conflicting_orders = score_counts[
    score_counts > 1
].index

conflicting_reviews = (
    reviews[reviews["order_id"].isin(conflicting_orders)]
    .sort_values(["order_id", "review_creation_date"])
)

print("\nCONFLICTING REVIEW SCORE DETAILS")
print("=" * 65)
print("Orders with conflicting scores:", len(conflicting_orders))

display(conflicting_reviews.head(30))

# ------------------------------------------------------------
# 4. Examine the score combinations across conflicting orders.
# ------------------------------------------------------------

conflict_score_patterns = (
    conflicting_reviews.groupby("order_id")["review_score"]
    .apply(lambda x: " / ".join(map(str, sorted(x.unique()))))
    .value_counts()
)

print("\nCONFLICTING SCORE COMBINATIONS")
print("=" * 65)
display(conflict_score_patterns)

# ------------------------------------------------------------
# 5. Check whether conflicting reviews have different
#    comments or creation dates.
# ------------------------------------------------------------

conflict_summary = (
    conflicting_reviews.groupby("order_id")
    .agg(
        number_of_reviews=("review_id", "size"),
        number_of_scores=("review_score", "nunique"),
        number_of_creation_dates=("review_creation_date", "nunique"),
        number_of_comments=("review_comment_message", "nunique")
    )
)

print("\nCONFLICTING ORDER CHARACTERISTICS")
print("=" * 65)
display(conflict_summary.describe())
```

```python

# ============================================================
# COMPARE ORDER-LEVEL REVIEW TARGET DEFINITIONS
# ============================================================

reviews = datasets["reviews"].copy()

# Convert review creation date to datetime
reviews["review_creation_date"] = pd.to_datetime(
    reviews["review_creation_date"],
    errors="coerce"
)

# Sort reviews chronologically within each order
reviews_sorted = reviews.sort_values(
    ["order_id", "review_creation_date", "review_answer_timestamp"]
)

# Earliest review per order
earliest_reviews = (
    reviews_sorted
    .drop_duplicates(subset="order_id", keep="first")
    [["order_id", "review_score"]]
    .rename(columns={"review_score": "earliest_score"})
)

# Worst (lowest) review score per order
worst_reviews = (
    reviews.groupby("order_id")["review_score"]
    .min()
    .reset_index()
    .rename(columns={"review_score": "worst_score"})
)

# Highest review score per order (for comparison)
best_reviews = (
    reviews.groupby("order_id")["review_score"]
    .max()
    .reset_index()
    .rename(columns={"review_score": "best_score"})
)

# Combine the target definitions
target_comparison = (
    earliest_reviews
    .merge(worst_reviews, on="order_id")
    .merge(best_reviews, on="order_id")
)

# Create binary targets
target_comparison["earliest_target_1_2"] = (
    target_comparison["earliest_score"] <= 2
).astype(int)

target_comparison["worst_target_1_2"] = (
    target_comparison["worst_score"] <= 2
).astype(int)

target_comparison["earliest_target_1_3"] = (
    target_comparison["earliest_score"] <= 3
).astype(int)

target_comparison["worst_target_1_3"] = (
    target_comparison["worst_score"] <= 3
).astype(int)

# Display target distribution
print("ORDER-LEVEL TARGET COMPARISON")
print("=" * 65)

target_columns = [
    "earliest_target_1_2",
    "worst_target_1_2",
    "earliest_target_1_3",
    "worst_target_1_3"
]

for column in target_columns:
    print(f"\n{column}")
    print(target_comparison[column].value_counts().sort_index())
    print(
        "Positive rate:",
        round(target_comparison[column].mean() * 100, 2),
        "%"
    )

# Compare how many orders receive different labels
print("\nTARGET DISAGREEMENT")
print("=" * 65)

print(
    "Disagreement between earliest and worst score (1-2):",
    (
        target_comparison["earliest_target_1_2"]
        != target_comparison["worst_target_1_2"]
    ).sum()
)

print(
    "Disagreement between earliest and worst score (1-3):",
    (
        target_comparison["earliest_target_1_3"]
        != target_comparison["worst_target_1_3"]
    ).sum()
)

# Show examples of orders whose labels change
changed_labels = target_comparison[
    target_comparison["earliest_target_1_2"]
    != target_comparison["worst_target_1_2"]
]

print("\nORDERS WITH DIFFERENT LABELS")
display(changed_labels.head(20))
```

```python

# ============================================================
#  CLEAN REVIEW DATA AND CREATE ORDER-LEVEL TARGET
# ============================================================

# Work on a copy; preserve the original dataset
reviews_clean = datasets["reviews"].copy()

# Convert review dates to datetime
reviews_clean["review_creation_date"] = pd.to_datetime(
    reviews_clean["review_creation_date"],
    errors="coerce"
)

reviews_clean["review_answer_timestamp"] = pd.to_datetime(
    reviews_clean["review_answer_timestamp"],
    errors="coerce"
)

# Sort reviews chronologically within each order
reviews_clean = reviews_clean.sort_values(
    ["order_id", "review_creation_date", "review_answer_timestamp"],
    na_position="last"
)

# Keep the earliest review for each order
order_reviews = (
    reviews_clean
    .drop_duplicates(subset="order_id", keep="first")
    [["order_id", "review_score", "review_creation_date"]]
    .copy()
)

# Create primary and alternative target definitions
order_reviews["low_satisfaction"] = (
    order_reviews["review_score"] <= 2
).astype(int)

order_reviews["low_satisfaction_1_3"] = (
    order_reviews["review_score"] <= 3
).astype(int)

# Validate the result
print("CLEANED ORDER-LEVEL REVIEW DATA")
print("=" * 65)

print("Original review rows:", len(datasets["reviews"]))
print("Cleaned order-level review rows:", len(order_reviews))
print("Unique order IDs:", order_reviews["order_id"].nunique())
print(
    "Duplicate order IDs:",
    order_reviews["order_id"].duplicated().sum()
)

print("\nPRIMARY TARGET DISTRIBUTION")
display(order_reviews["low_satisfaction"].value_counts().sort_index())

print("\nALTERNATIVE TARGET DISTRIBUTION")
display(order_reviews["low_satisfaction_1_3"].value_counts().sort_index())

print("\nPRIMARY TARGET POSITIVE RATE:")
print(round(order_reviews["low_satisfaction"].mean() * 100, 2), "%")

# Save a copy of the order-level target data
PROCESSED_DATA_PATH = PROJECT_ROOT / "data" / "processed"
PROCESSED_DATA_PATH.mkdir(parents=True, exist_ok=True)

order_reviews.to_csv(
    PROCESSED_DATA_PATH / "order_level_review_targets.csv",
    index=False
)

print("\nSaved to:")
print(PROCESSED_DATA_PATH / "order_level_review_targets.csv")
```

```python
# ============================================================
#  PREPARE DELIVERED ORDERS AND MERGE TARGETS
# ============================================================

# Work on a copy of the orders dataset
orders_clean = datasets["orders"].copy()

# Convert order date columns to datetime
date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for column in date_columns:
    orders_clean[column] = pd.to_datetime(
        orders_clean[column],
        errors="coerce"
    )

# Check delivery information for delivered orders
delivered_orders = orders_clean[
    orders_clean["order_status"] == "delivered"
].copy()

print("DELIVERED ORDER CHECK")
print("=" * 60)

print("Total delivered orders:", len(delivered_orders))

print(
    "Missing carrier delivery dates:",
    delivered_orders["order_delivered_carrier_date"].isna().sum()
)

print(
    "Missing customer delivery dates:",
    delivered_orders["order_delivered_customer_date"].isna().sum()
)

# Merge delivered orders with the order-level review targets
model_base = delivered_orders.merge(
    order_reviews,
    on="order_id",
    how="inner",
    validate="one_to_one"
)

# Check the merged dataset
print("\nDELIVERED ORDERS WITH REVIEW TARGETS")
print("=" * 60)

print("Delivered orders with reviews:", len(model_base))
print("Unique order IDs:", model_base["order_id"].nunique())
print("Duplicate order IDs:", model_base["order_id"].duplicated().sum())

print("\nPRIMARY TARGET DISTRIBUTION")
display(model_base["low_satisfaction"].value_counts().sort_index())

print("\nPRIMARY TARGET POSITIVE RATE:")
print(round(model_base["low_satisfaction"].mean() * 100, 2), "%")

# Check how many delivered orders have no review target
orders_without_reviews = len(delivered_orders) - len(model_base)

print("\nDelivered orders without reviews:", orders_without_reviews)

# Save the intermediate modelling dataset
model_base.to_csv(
    PROCESSED_DATA_PATH / "delivered_orders_with_targets.csv",
    index=False
)

print("\nSaved to:")
print(PROCESSED_DATA_PATH / "delivered_orders_with_targets.csv")
```

```python
# ============================================================
# SECTION 7: INVESTIGATE MISSING DELIVERY DATES
# ============================================================

# Identify delivered orders with missing actual delivery dates
missing_delivery = model_base[
    model_base["order_delivered_customer_date"].isna()
].copy()

missing_carrier = model_base[
    model_base["order_delivered_carrier_date"].isna()
].copy()

print("MISSING CUSTOMER DELIVERY DATE")
print("=" * 60)

display(
    missing_delivery[
        [
            "order_id",
            "order_status",
            "order_purchase_timestamp",
            "order_delivered_carrier_date",
            "order_delivered_customer_date",
            "order_estimated_delivery_date",
            "review_score",
            "low_satisfaction"
        ]
    ]
)

print("\nMISSING CARRIER DELIVERY DATE")
print("=" * 60)

display(
    missing_carrier[
        [
            "order_id",
            "order_status",
            "order_purchase_timestamp",
            "order_delivered_carrier_date",
            "order_delivered_customer_date",
            "order_estimated_delivery_date",
            "review_score",
            "low_satisfaction"
        ]
    ]
)

print("\nMISSING DATE SUMMARY")
print("Orders missing customer delivery date:", len(missing_delivery))
print("Orders missing carrier delivery date:", len(missing_carrier))
```

```python
# ============================================================
#  CREATE CLEAN MODELLING BASE
# ============================================================

# Exclude orders without actual customer delivery timestamps
model_base_clean = model_base.dropna(
    subset=["order_delivered_customer_date"]
).copy()

# Calculate delivery duration
model_base_clean["delivery_days"] = (
    model_base_clean["order_delivered_customer_date"]
    - model_base_clean["order_purchase_timestamp"]
).dt.total_seconds() / 86400

# Calculate delivery delay against the estimated delivery date
model_base_clean["delivery_delay_days"] = (
    model_base_clean["order_delivered_customer_date"]
    - model_base_clean["order_estimated_delivery_date"]
).dt.total_seconds() / 86400

# Create an indicator for late delivery
model_base_clean["late_delivery"] = (
    model_base_clean["delivery_delay_days"] > 0
).astype(int)

# Validate the cleaned dataset
print("CLEAN MODELLING BASE")
print("=" * 60)

print("Original delivered orders with reviews:", len(model_base))
print("Excluded orders with missing delivery dates:", len(model_base) - len(model_base_clean))
print("Remaining modelling records:", len(model_base_clean))
print("Unique order IDs:", model_base_clean["order_id"].nunique())
print("Duplicate order IDs:", model_base_clean["order_id"].duplicated().sum())

print("\nTARGET DISTRIBUTION")
display(
    model_base_clean["low_satisfaction"]
    .value_counts()
    .sort_index()
)

print("\nTARGET POSITIVE RATE:")
print(round(model_base_clean["low_satisfaction"].mean() * 100, 2), "%")

print("\nDELIVERY FEATURE SUMMARY")
display(
    model_base_clean[
        ["delivery_days", "delivery_delay_days", "late_delivery"]
    ].describe()
)

print("\nLATE DELIVERY DISTRIBUTION")
display(model_base_clean["late_delivery"].value_counts())

# Save the clean modelling base
model_base_clean.to_csv(
    PROCESSED_DATA_PATH / "clean_modelling_base.csv",
    index=False
)

print("\nSaved to:")
print(PROCESSED_DATA_PATH / "clean_modelling_base.csv")
```

```python
# ============================================================
#  CALCULATE CUSTOMER-TO-SELLER DISTANCES
# ============================================================

import numpy as np

# Connect each order to its customer
order_customer = orders_clean[
    ["order_id", "customer_id"]
].drop_duplicates(subset="order_id")

# Attach customer IDs and coordinates to order items
distance_data = items_clean.merge(
    order_customer,
    on="order_id",
    how="left",
    validate="many_to_one"
)

distance_data = distance_data.merge(
    customers_geo[
        ["customer_id", "customer_latitude", "customer_longitude"]
    ],
    on="customer_id",
    how="left",
    validate="many_to_one"
)

# Attach seller coordinates
distance_data = distance_data.merge(
    sellers_geo[
        ["seller_id", "seller_latitude", "seller_longitude"]
    ],
    on="seller_id",
    how="left",
    validate="many_to_one"
)

# Keep one record per order and seller
distance_data = distance_data.drop_duplicates(
    subset=["order_id", "seller_id"]
).copy()

# Calculate Haversine distance
def haversine_distance(lat1, lon1, lat2, lon2):
    earth_radius_km = 6371

    lat1 = np.radians(lat1)
    lon1 = np.radians(lon1)
    lat2 = np.radians(lat2)
    lon2 = np.radians(lon2)

    latitude_difference = lat2 - lat1
    longitude_difference = lon2 - lon1

    a = (
        np.sin(latitude_difference / 2) ** 2
        + np.cos(lat1) * np.cos(lat2)
        * np.sin(longitude_difference / 2) ** 2
    )

    return 2 * earth_radius_km * np.arcsin(
        np.sqrt(np.clip(a, 0, 1))
    )

# Identify records with complete coordinate information
valid_distance = distance_data[
    [
        "customer_latitude",
        "customer_longitude",
        "seller_latitude",
        "seller_longitude"
    ]
].notna().all(axis=1)

# Calculate distance for valid coordinate pairs
distance_data["customer_seller_distance_km"] = np.nan

distance_data.loc[valid_distance, "customer_seller_distance_km"] = (
    haversine_distance(
        distance_data.loc[valid_distance, "customer_latitude"],
        distance_data.loc[valid_distance, "customer_longitude"],
        distance_data.loc[valid_distance, "seller_latitude"],
        distance_data.loc[valid_distance, "seller_longitude"]
    )
)

# Aggregate distances at order level
order_distance_features = (
    distance_data
    .groupby("order_id")
    .agg(
        average_seller_distance_km=(
            "customer_seller_distance_km", "mean"
        ),
        maximum_seller_distance_km=(
            "customer_seller_distance_km", "max"
        ),
        sellers_with_coordinates=(
            "customer_seller_distance_km", "count"
        )
    )
    .reset_index()
)

# Validate the results
print("CUSTOMER-TO-SELLER DISTANCE FEATURES")
print("=" * 60)

print("Unique order-seller pairs:", len(distance_data))
print("Orders with distance features:", len(order_distance_features))
print("Valid distance records:", valid_distance.sum())
print(
    "Orders without valid distances:",
    order_distance_features["average_seller_distance_km"].isna().sum()
)

print("\nDISTANCE SUMMARY (KM)")
display(
    order_distance_features[
        [
            "average_seller_distance_km",
            "maximum_seller_distance_km"
        ]
    ].describe()
)

# Save order-level distance features
order_distance_features.to_csv(
    PROCESSED_DATA_PATH / "order_distance_features.csv",
    index=False
)

print("\nSaved corrected order-level distance features successfully.")
```

```python
# ============================================================
#  CREATE HISTORICAL CUSTOMER FEATURES
# ============================================================

# Prepare customer order history
customer_history = full_integrated_data[
    [
        "order_id",
        "customer_unique_id",
        "order_purchase_timestamp",
        "review_score"
    ]
].copy()

# Ensure purchase timestamps are datetime values
customer_history["order_purchase_timestamp"] = pd.to_datetime(
    customer_history["order_purchase_timestamp"],
    errors="coerce"
)

# Sort customer orders chronologically
customer_history = customer_history.sort_values(
    ["customer_unique_id", "order_purchase_timestamp", "order_id"]
).copy()

# Count previous orders before the current order
customer_history["customer_previous_orders"] = (
    customer_history.groupby("customer_unique_id").cumcount()
)

# Calculate historical satisfaction using previous orders only
customer_history["customer_previous_avg_review"] = (
    customer_history.groupby("customer_unique_id")["review_score"]
    .transform(lambda x: x.shift(1).expanding().mean())
)

# Create order-level customer history features
customer_history_features = customer_history[
    [
        "order_id",
        "customer_previous_orders",
        "customer_previous_avg_review"
    ]
].copy()

# Validate the results
print("CUSTOMER HISTORICAL FEATURES")
print("=" * 60)

print("Total records:", len(customer_history_features))
print(
    "Duplicate order IDs:",
    customer_history_features["order_id"].duplicated().sum()
)

print("\nPREVIOUS ORDER COUNT")
display(
    customer_history_features["customer_previous_orders"].describe()
)

print("\nPREVIOUS AVERAGE REVIEW SCORE")
display(
    customer_history_features["customer_previous_avg_review"].describe()
)

print("\nMissing historical average review values:")
print(
    customer_history_features[
        "customer_previous_avg_review"
    ].isna().sum()
)

# Save customer historical features
customer_history_features.to_csv(
    PROCESSED_DATA_PATH / "customer_history_features.csv",
    index=False
)

print("\nCustomer historical features saved successfully.")
```

```python
# ============================================================
#  CREATE HISTORICAL SELLER PERFORMANCE FEATURES
# ============================================================

# Prepare seller order history
seller_history = distance_data[
    ["order_id", "seller_id"]
].drop_duplicates().merge(
    full_integrated_data[
        [
            "order_id",
            "order_purchase_timestamp",
            "delivery_delay_days",
            "low_satisfaction"
        ]
    ],
    on="order_id",
    how="left",
    validate="many_to_one"
)

# Convert purchase timestamps to datetime
seller_history["order_purchase_timestamp"] = pd.to_datetime(
    seller_history["order_purchase_timestamp"],
    errors="coerce"
)

# Sort seller orders chronologically
seller_history = seller_history.sort_values(
    ["seller_id", "order_purchase_timestamp", "order_id"]
).copy()

# Count previous orders handled by each seller
seller_history["seller_previous_orders"] = (
    seller_history.groupby("seller_id").cumcount()
)

# Calculate historical average delivery delay using previous orders
seller_history["seller_previous_avg_delay"] = (
    seller_history.groupby("seller_id")["delivery_delay_days"]
    .transform(lambda x: x.shift(1).expanding().mean())
)

# Calculate historical low-satisfaction rate using previous orders
seller_history["seller_previous_low_satisfaction_rate"] = (
    seller_history.groupby("seller_id")["low_satisfaction"]
    .transform(lambda x: x.shift(1).expanding().mean())
)

# Aggregate seller history features at order level
seller_history_features = (
    seller_history
    .groupby("order_id")
    .agg(
        average_seller_previous_orders=(
            "seller_previous_orders", "mean"
        ),
        average_seller_previous_delay=(
            "seller_previous_avg_delay", "mean"
        ),
        average_seller_previous_low_satisfaction_rate=(
            "seller_previous_low_satisfaction_rate", "mean"
        )
    )
    .reset_index()
)

# Validate the results
print("HISTORICAL SELLER FEATURES")
print("=" * 60)

print("Seller-order records:", len(seller_history))
print("Order-level records:", len(seller_history_features))
print(
    "Duplicate order IDs:",
    seller_history_features["order_id"].duplicated().sum()
)

print("\nPREVIOUS SELLER ORDER COUNT")
display(
    seller_history_features[
        "average_seller_previous_orders"
    ].describe()
)

print("\nPREVIOUS SELLER DELIVERY DELAY")
display(
    seller_history_features[
        "average_seller_previous_delay"
    ].describe()
)

print("\nPREVIOUS LOW-SATISFACTION RATE")
display(
    seller_history_features[
        "average_seller_previous_low_satisfaction_rate"
    ].describe()
)

# Save seller historical features
seller_history_features.to_csv(
    PROCESSED_DATA_PATH / "seller_history_features.csv",
    index=False
)

print("\nSeller historical features saved successfully.")
```

```python
# ============================================================
# CREATE LEAKAGE-SAFE PRODUCT HISTORICAL FEATURES
# ============================================================

# Prepare product-order history
product_history = items_clean[
    ["order_id", "product_id"]
].drop_duplicates().merge(
    full_integrated_data[
        [
            "order_id",
            "order_delivered_customer_date",
            "delivery_delay_days"
        ]
    ],
    on="order_id",
    how="inner",
    validate="many_to_one"
)

# Ensure delivery timestamps are datetime values
product_history["order_delivered_customer_date"] = pd.to_datetime(
    product_history["order_delivered_customer_date"],
    errors="coerce"
)

# Aggregate products delivered at the same timestamp
product_timestamp_history = (
    product_history
    .groupby(
        ["product_id", "order_delivered_customer_date"],
        as_index=False
    )
    .agg(
        completed_product_orders=("order_id", "nunique"),
        average_delivery_delay=("delivery_delay_days", "mean")
    )
    .sort_values(
        ["product_id", "order_delivered_customer_date"]
    )
)

# Calculate historical values using strictly earlier delivery timestamps
product_timestamp_history["product_previous_completed_orders"] = (
    product_timestamp_history.groupby("product_id")[
        "completed_product_orders"
    ].cumsum()
    - product_timestamp_history["completed_product_orders"]
)

product_timestamp_history["product_previous_avg_delay"] = (
    product_timestamp_history.groupby("product_id")[
        "average_delivery_delay"
    ]
    .transform(
        lambda x: (
            x.shift(1).expanding().mean()
        )
    )
)

# Attach historical product performance to each product-order record
product_history = product_history.merge(
    product_timestamp_history[
        [
            "product_id",
            "order_delivered_customer_date",
            "product_previous_completed_orders",
            "product_previous_avg_delay"
        ]
    ],
    on=["product_id", "order_delivered_customer_date"],
    how="left",
    validate="many_to_one"
)

# Aggregate product history at order level
product_history_features = (
    product_history
    .groupby("order_id")
    .agg(
        average_product_previous_completed_orders=(
            "product_previous_completed_orders", "mean"
        ),
        average_product_previous_delay=(
            "product_previous_avg_delay", "mean"
        )
    )
    .reset_index()
)

# Validate the results
print("LEAKAGE-SAFE PRODUCT HISTORICAL FEATURES")
print("=" * 60)

print("Product-order records:", len(product_history))
print("Order-level records:", len(product_history_features))
print(
    "Duplicate order IDs:",
    product_history_features["order_id"].duplicated().sum()
)

print("\nPREVIOUS COMPLETED PRODUCT ORDERS")
display(
    product_history_features[
        "average_product_previous_completed_orders"
    ].describe()
)

print("\nPREVIOUS PRODUCT DELIVERY DELAY")
display(
    product_history_features[
        "average_product_previous_delay"
    ].describe()
)

# Save product historical features
product_history_features.to_csv(
    PROCESSED_DATA_PATH / "product_history_features.csv",
    index=False
)

print("\nProduct historical features saved successfully.")
```

```python
# ============================================================
# DEFINE MODELLING FEATURES AND EXCLUDE LEAKAGE
# ============================================================

# Target variables
target_columns = [
    "low_satisfaction",
    "low_satisfaction_1_3"
]

# Identifiers and direct order references
identifier_columns = [
    "order_id",
    "customer_id",
    "customer_unique_id"
]

# Review information that must not be used as predictors
review_columns = [
    "review_score",
    "review_creation_date"
]

# Fields unavailable or unnecessary at the prediction point
excluded_columns = [
    "order_status",
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date",
    "payment_methods",
    "customer_city",
    "customer_zip_code_prefix",
    "product_categories",
    "seller_states"
]

# Combine all columns excluded from model predictors
excluded_from_features = (
    target_columns
    + identifier_columns
    + review_columns
    + excluded_columns
)

# Select predictor columns
feature_columns = [
    column
    for column in full_integrated_data.columns
    if column not in excluded_from_features
]

# Display the feature selection results
print("MODELLING FEATURE SELECTION")
print("=" * 60)

print("Total dataset columns:", len(full_integrated_data.columns))
print("Excluded columns:", len(excluded_from_features))
print("Selected predictor columns:", len(feature_columns))

print("\nSelected predictors:")
for column in feature_columns:
    print(column)

print("\nExcluded columns:")
for column in excluded_from_features:
    if column in full_integrated_data.columns:
        print(column)

# Confirm that no target or review information remains in predictors
assert "low_satisfaction" not in feature_columns
assert "low_satisfaction_1_3" not in feature_columns
assert "review_score" not in feature_columns
assert "review_creation_date" not in feature_columns

print("\nLeakage checks passed.")
```

```python
# ============================================================
# INSPECT SELECTED FEATURE DATA TYPES
# ============================================================

# Remove the payment indicator with no variation
feature_columns = [
    column for column in feature_columns
    if column != "used_not_defined"
]

# Create a feature inspection table
feature_inspection = pd.DataFrame({
    "feature": feature_columns,
    "data_type": [
        full_integrated_data[column].dtype
        for column in feature_columns
    ],
    "missing_values": [
        full_integrated_data[column].isna().sum()
        for column in feature_columns
    ],
    "unique_values": [
        full_integrated_data[column].nunique(dropna=False)
        for column in feature_columns
    ]
})

print("FEATURE DATA TYPE INSPECTION")
print("=" * 60)

display(feature_inspection)

print("\nFEATURES BY DATA TYPE:")
display(
    feature_inspection["data_type"]
    .astype(str)
    .value_counts()
)

# Identify constant features
constant_features = [
    column for column in feature_columns
    if full_integrated_data[column].nunique(dropna=False) <= 1
]

print("\nCONSTANT FEATURES:")
print(constant_features)

print("\nFinal predictor count:", len(feature_columns))
```

```python
# ============================================================
# PREPARE THE MODELLING DATASET
# ============================================================

# Create the predictor dataset
X = full_integrated_data[feature_columns].copy()

# Create the primary target
y = full_integrated_data["low_satisfaction"].copy()

# Retain identifiers and timestamps for chronological splitting
model_metadata = full_integrated_data[
    [
        "order_id",
        "order_purchase_timestamp"
    ]
].copy()

# Ensure purchase timestamps are datetime values
model_metadata["order_purchase_timestamp"] = pd.to_datetime(
    model_metadata["order_purchase_timestamp"],
    errors="coerce"
)

# Sort all modelling records chronologically
chronological_order = (
    model_metadata["order_purchase_timestamp"]
    .sort_values()
    .index
)

X = X.loc[chronological_order].reset_index(drop=True)
y = y.loc[chronological_order].reset_index(drop=True)
model_metadata = (
    model_metadata.loc[chronological_order]
    .reset_index(drop=True)
)

# Validate the modelling dataset
print("FINAL MODELLING DATASET")
print("=" * 60)

print("Number of observations:", len(X))
print("Number of predictors:", X.shape[1])
print("Target observations:", len(y))

print("\nTarget distribution:")
display(
    y.value_counts()
    .rename(index={0: "Not low satisfaction", 1: "Low satisfaction"})
)

print("\nTarget percentage:")
display(
    (y.value_counts(normalize=True) * 100)
    .rename(index={0: "Not low satisfaction", 1: "Low satisfaction"})
)

print("\nPredictor data types:")
display(X.dtypes.value_counts())

print("\nMissing predictor values:")
display(
    X.isna().sum()
    .loc[lambda values: values > 0]
)

print("\nChronological date range:")
print("Start:", model_metadata["order_purchase_timestamp"].min())
print("End:", model_metadata["order_purchase_timestamp"].max())

# Confirm alignment and leakage exclusions
assert len(X) == len(y) == len(model_metadata)
assert "low_satisfaction" not in X.columns
assert "review_score" not in X.columns
assert X.columns.tolist() == feature_columns

print("\nModelling dataset prepared successfully.")
```
